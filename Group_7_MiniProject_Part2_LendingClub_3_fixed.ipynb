## 1. Problem Definition

### 1.1 ปัญหา/คำถามที่สนใจ
**คำถาม:** ความน่าจะเป็นที่ลูกค้าแต่ละรายมีโอกาสผิดนัดชำระหนี้ โดยพิจารณาจากข้อมูล ณ วันที่สมัครขอสินเชื่อ

* **ผลลัพธ์ที่ต้องการ:** ค่า **ความน่าจะเป็นผิดนัด (Probability of Default, PD)** ของลูกค้าแต่ละราย เช่น "ลูกค้ารายนี้มีโอกาสผิดนัด 27%" — ไม่ใช่แค่ทายว่า ผิดนัด/ไม่ผิดนัด
* **"ข้อมูล ณ วันที่สมัคร"** = ข้อมูลในใบสมัครและรายงานเครดิต (รายได้, อายุงาน, FICO, DTI, ประวัติค้างชำระ ฯลฯ) และเงื่อนไขสินเชื่อ ณ วันอนุมัติ (วงเงิน, ระยะเวลา, เกรด, ดอกเบี้ย) — **ไม่ใช้ข้อมูลที่เกิดขึ้นหลังอนุมัติ** เช่น ยอดชำระ เงินที่ติดตามคืน (ป้องกัน data leakage)

### 1.2 ประเภทปัญหา: **Classification (Binary, แบบให้ผลเป็นความน่าจะเป็น)**

| | รายละเอียด |
|---|---|
| ประเภทปัญหา | **Binary Classification** (Supervised learning) — ใช้ `predict_proba` ของโมเดลเป็น % ผิดนัด |
| เหตุผลที่ไม่ใช่ Regression | target จริงมีแค่ 2 ค่า (ผิดนัด = 1 / ชำระครบ = 0) ไม่มี "% ผิดนัด" ของแต่ละคนให้เรียนรู้โดยตรง — classifier ที่ให้ความน่าจะเป็นจึงเป็นวิธีมาตรฐานในการประมาณ PD |
| Target | `default` สร้างจาก `loan_status`: **1** = Charged Off / Default, **0** = Fully Paid (ตัดสถานะที่ยังไม่สิ้นสุด เช่น Current, Late ออก) |
| Output | PD (%) ของลูกค้าแต่ละราย ที่ผ่านการ **calibrate** ให้ตรงกับอัตราผิดนัดจริง + คำแนะนำ อนุมัติ / ตรวจสอบเพิ่ม ตาม threshold |
| โมเดล | เปรียบเทียบ **9 อัลกอริทึม** (+ baseline) แล้ว tune 3 อันดับแรก |

### 1.2 Business Value & Impact
1. Reduce NPLs: ควบคุมและลดอัตราหนี้เสียด้วยระบบคัดกรองความเสี่ยงล่วงหน้า (Predictive Power)
2. Accelerate Time-to-Market: เร่งสปีดการพิจารณาสินเชื่อให้เร็วขึ้น (Auto-Approval) เพิ่มโอกาสปิดยอดขาย
3. Optimize Human Resources: ลดภาระงานเอกสารของเจ้าหน้าที่ เพื่อให้ทีม Underwriter โฟกัสเฉพาะเคสที่ต้อง "ตรวจสอบเพิ่ม" (Manual Review) ได้อย่างเต็มประสิทธิภาพ

### 1.3 Model Metric (จะใช้อะไรวัดผล)
คำถามต้องการ "%" จึงต้องวัด 2 ด้าน: **(ก) จัดลำดับความเสี่ยงถูกไหม** และ **(ข) % ที่ทำนายตรงกับความจริงไหม**

| Metric | ด้าน | บทบาท | เหตุผล |
|---|---|---|---|
| **ROC-AUC** | (ก) Discrimination | metric หลักใน model selection / tuning | วัดว่าลูกค้าที่ผิดนัดได้ % สูงกว่าลูกค้าที่ชำระครบหรือไม่ ไม่ขึ้นกับ threshold ไม่ถูกหลอกโดย imbalance (0.5 = เดาสุ่ม) |
| **PR-AUC** | (ก) | เลือกวิธีจัดการ imbalance | เน้นคลาสส่วนน้อย (ผิดนัด) เทียบกับค่าเดาสุ่ม = อัตราผิดนัด |
| **Brier score / Log loss** | (ข) Calibration | วัดความแม่นของ % | Brier = ค่าเฉลี่ยของ (PD − ผลจริง)² ยิ่งต่ำยิ่งดี |
| **Calibration curve + MAE / RMSE (จุด%)** ระหว่าง % ที่ทำนายกับอัตราผิดนัดจริงรายกลุ่ม (decile) | (ข) | ตรวจว่า "ลูกค้าที่โมเดลบอก 30%" ผิดนัดจริงราว 30% หรือไม่ | Error อยู่ในหน่วยเดียวกับคำตอบ (จุด%) เข้าใจง่ายสำหรับผู้บริหาร |
| **Precision, Recall, F1** (จาก confusion matrix) | การตัดสินใจ | เมื่อแปลง % เป็นการอนุมัติ/ปฏิเสธด้วย threshold | **Recall** สำคัญเพราะ FN เสียหายมากกว่า FP; threshold = **จุดคุ้มทุนของ PD** ที่คำนวณจากกำไร/ขาดทุนจริงใน train (หัวข้อ 10.3) |
| **กำไรจริง / ขาดทุนที่หลีกเลี่ยงได้ ($)** | Business metric | เทียบนโยบายใช้โมเดลกับ "อนุมัติทุกราย" | ผู้บริหารเข้าใจง่ายที่สุด ตอบ Business Goal (ลดต้นทุนหนี้เสีย) โดยตรง |
| ~~Accuracy~~ | – | ไม่ใช้ | ข้อมูลไม่สมดุล ทายว่า "ไม่ผิดนัด" ทุกรายก็ได้ accuracy ~80% |

| `fico_range_low`, `fico_range_high` | numeric | ช่วงคะแนน FICO ณ วันยื่นกู้ |
| `inq_last_6mths` | numeric | จำนวนการสอบถามเครดิตใน 6 เดือน |

```python
# ตรวจสอบ Data Dictionary: คอลัมน์ครบหรือไม่ / ชนิดข้อมูลตรงหรือไม่ / ตัวอย่างค่า / missing
DATA_DICT = {
    'id': 'id', 'loan_status': 'category', 'issue_d': 'date', 'loan_amnt': 'numeric', 'term': 'category',
    'int_rate': 'numeric', 'installment': 'numeric', 'grade': 'category', 'sub_grade': 'category',
    'emp_length': 'category', 'home_ownership': 'category', 'annual_inc': 'numeric',
    'verification_status': 'category', 'purpose': 'category', 'addr_state': 'category', 'dti': 'numeric',
    'delinq_2yrs': 'numeric', 'earliest_cr_line': 'date', 'fico_range_low': 'numeric', 'fico_range_high': 'numeric',
    'inq_last_6mths': 'numeric', 'open_acc': 'numeric', 'pub_rec': 'numeric', 'revol_bal': 'numeric',
    'revol_util': 'numeric', 'total_acc': 'numeric', 'mort_acc': 'numeric', 'pub_rec_bankruptcies': 'numeric',
    'application_type': 'category', 'initial_list_status': 'category',
    'total_pymnt': 'numeric (business evaluation only)',
}
missing_cols = [c for c in DATA_DICT if c not in raw.columns]
print('คอลัมน์ใน dictionary ที่ไม่พบในไฟล์:', missing_cols if missing_cols else 'ไม่มี (ครบทุกคอลัมน์) ✔')

def _type_ok(col, expected):
    s = raw[col]
    if expected.startswith('numeric'): return pd.api.types.is_numeric_dtype(s)
    return not pd.api.types.is_numeric_dtype(s)        # id / category / date อ่านเป็นข้อความ (แปลงชนิดในข้อ 2.3)

dd = pd.DataFrame({
    'ชนิดตาม dictionary': pd.Series(DATA_DICT),
    'dtype ที่อ่านได้': raw.dtypes.astype(str),
    'ชนิดตรงกัน': pd.Series({c: _type_ok(c, t) for c, t in DATA_DICT.items()}),
    'ตัวอย่างค่า': raw.apply(lambda s: s.dropna().iloc[0] if s.notna().any() else None),
    'n_unique': raw.nunique(),
    'n_missing': raw.isna().sum(),
    'pct_missing': raw.isna().mean() * 100,
}).loc[list(DATA_DICT)]
display(dd)
print(f'Memory usage: {raw.memory_usage(deep=True).sum()/1024**2:,.1f} MB')

# ข้อมูลซ้ำ และแถวที่ไม่ใช่สินเชื่อ
n_footer = raw['loan_amnt'].isna().sum()
print(f'แถวที่ไม่มี loan_amnt (แถวสรุปท้ายไฟล์/ข้อมูลไม่สมบูรณ์): {n_footer:,}')
print(f'แถวซ้ำทั้งแถว: {raw.duplicated().sum():,}')
print(f'id ซ้ำ (ไม่นับค่าว่าง): {raw["id"].dropna().duplicated().sum():,}')

# ค่าที่เป็นไปไม่ได้ / ไม่สอดคล้องกัน (จัดการในหัวข้อ 4)
issue_dt = pd.to_datetime(raw['issue_d'].astype(str), format='%b-%Y', errors='coerce')
ecl_dt = pd.to_datetime(raw['earliest_cr_line'].astype(str), format='%b-%Y', errors='coerce')
invalid_checks = {
    'annual_inc <= 0': (raw['annual_inc'] <= 0).sum(),
    'dti < 0': (raw['dti'] < 0).sum(),
    'dti > 100 (เช่น 999 = ค่า placeholder)': (raw['dti'] > 100).sum(),
    'revol_util > 150%': (raw['revol_util'] > 150).sum(),
    'int_rate <= 0 หรือ > 40%': ((raw['int_rate'] <= 0) | (raw['int_rate'] > 40)).sum(),
    'installment > loan_amnt': (raw['installment'] > raw['loan_amnt']).sum(),
    'fico_range_low > fico_range_high': (raw['fico_range_low'] > raw['fico_range_high']).sum(),
    'earliest_cr_line หลัง issue_d': (ecl_dt > issue_dt).sum(),
    'open_acc > total_acc': (raw['open_acc'] > raw['total_acc']).sum(),
}
display(pd.Series(invalid_checks, name='จำนวนแถว').to_frame())

# สถานะสินเชื่อ (ใช้กำหนด target)
display(raw['loan_status'].value_counts(dropna=False).to_frame('n'))
```

```python
# ตรวจสอบจำนวนและเปอร์เซ็นต์ของค่าว่าง (Missing Values) ในข้อมูลดิบ (raw)
# หมายเหตุ: หัวข้อ 2 ยังไม่ได้สร้าง train_df / TARGET จึงตรวจจาก raw และสร้าง target ชั่วคราวจาก loan_status
missing_data = pd.DataFrame({
    'จำนวนค่าว่าง (n_missing)': raw.isna().sum(),
    'เปอร์เซ็นต์ค่าว่าง (%)': raw.isna().mean() * 100
}).sort_values('จำนวนค่าว่าง (n_missing)', ascending=False)

print("=== Data Quality: Missing Values Assessment ===")
display(missing_data.head(10))

print("\n=== Data Quality: Class Imbalance Assessment ===")
_status_map = {'Fully Paid': 0, 'Does not meet the credit policy. Status:Fully Paid': 0,
               'Charged Off': 1, 'Default': 1, 'Does not meet the credit policy. Status:Charged Off': 1}
_y = raw['loan_status'].astype(str).map(_status_map).dropna().astype(int)   # เฉพาะสินเชื่อที่สิ้นสุดแล้ว
target_counts = _y.value_counts().sort_index()
target_pct = _y.value_counts(normalize=True).sort_index() * 100

imbalance_check = pd.DataFrame({
    'จำนวนแถว (Count)': target_counts,
    'สัดส่วน (Percentage)': target_pct
}, index=[0, 1])
imbalance_check.index.name = 'Class (default)'
imbalance_check.rename(index={0: 'Fully Paid (0)', 1: 'Charged Off / Default (1)'}, inplace=True)

display(imbalance_check)

ratio = target_counts[0] / target_counts[1]
print(f"อัตราส่วนความไม่สมดุล (Imbalance Ratio): Fully Paid ต่อ Charged Off = {ratio:.2f} : 1")
if ratio > 3:
    print("สถานะ: ข้อมูลมีความไม่สมดุลสูงอย่างมีนัยสำคัญ (Highly Imbalanced) จำเป็นต้องใช้เทคนิคพิเศษในการจัดการ เช่น SMOTE")
else:
    print("สถานะ: ข้อมูลค่อนข้างสมดุลดี")
del _y
```

### 2.3 กำหนดขอบเขตข้อมูลและสร้าง Target
ขั้นตอนนี้เป็นการจัดการ missing / outlier / encoding / scaling ที่ต้องเรียนรู้จากข้อมูล มีขั้นตอน ดังนี้

| ขั้นตอน | สิ่งที่ทำ | เหตุผล |
|---|---|---|
| 0 | ข้อมูลตั้งต้น | – |
| 1 | ลบแถวที่ไม่มี `loan_amnt` / `issue_d` | เป็นแถวสรุปท้ายไฟล์ ไม่ใช่สินเชื่อ |
| 2 | ลบ `id` ซ้ำ (เก็บแถวแรก) | สินเชื่อเดียวกันไม่ควรถูกนับ 2 ครั้ง (อาจไปอยู่ทั้ง train และ test) |
| 3 | เก็บเฉพาะสถานะสิ้นสุด (Fully Paid / Charged Off / Default) และสร้าง `default` | สินเชื่อ Current / Late ยังไม่รู้ผลสุดท้าย ใส่ label ไม่ได้ |
| 4 | เก็บเฉพาะสินเชื่อที่ออกปี 2007–2016 | สินเชื่อปี 2017–2018 ส่วนใหญ่ยังไม่ครบกำหนด ส่วนที่ปิดแล้วจึงเอนไปทาง "ผิดนัดเร็ว/ปิดก่อนกำหนด" (right-censoring) ทำให้ target มี bias |
| 5 | แปลงชนิดข้อมูล: วันที่ → datetime, `term` → จำนวนเดือน, `emp_length` → จำนวนปี | ข้อความที่จริง ๆ เป็นตัวเลข/วันที่ ต้องแปลงก่อนวิเคราะห์ |

(จำนวนแถวที่เปลี่ยนแปลงในแต่ละขั้นอยู่ในตารางที่โค้ดพิมพ์ออกมา)

```python
clean_log = []
def log_step(step, action, df_):
    clean_log.append({'ขั้นตอน': step, 'สิ่งที่ทำ': action, 'แถวคงเหลือ': len(df_),
                      'แถวที่ลดลง': (clean_log[-1]['แถวคงเหลือ'] - len(df_)) if clean_log else 0})

# ตรวจสอบหากไม่มีตัวแปร 'raw' ในระบบ ให้ทำการโหลดใหม่เพื่อป้องกัน NameError
if 'raw' not in globals():
    print('ไม่พบตัวแปร raw ในระบบ ทำการโหลดข้อมูลใหม่จากไฟล์...')
    raw = pd.read_csv(DATA_FILE, usecols=USECOLS, dtype=CAT_DTYPES, low_memory=False)

df = raw.copy()
log_step(0, 'ข้อมูลตั้งต้น', df)

df = df[df['loan_amnt'].notna() & df['issue_d'].notna()]
log_step(1, 'ลบแถวที่ไม่มี loan_amnt / issue_d', df)

df = df[~(df['id'].notna() & df['id'].duplicated(keep='first'))]
log_step(2, 'ลบ id ซ้ำ', df)

STATUS_MAP = {
    'Fully Paid': 0, 'Does not meet the credit policy. Status:Fully Paid': 0,
    'Charged Off': 1, 'Default': 1, 'Does not meet the credit policy. Status:Charged Off': 1,
}
df['default'] = df['loan_status'].astype(str).map(STATUS_MAP)
df = df[df['default'].notna()].copy()
df['default'] = df['default'].astype(int)
log_step(3, 'เก็บเฉพาะสินเชื่อที่สิ้นสุดแล้ว + สร้าง target', df)

df['issue_d'] = pd.to_datetime(df['issue_d'].astype(str), format='%b-%Y')
df['earliest_cr_line'] = pd.to_datetime(df['earliest_cr_line'].astype(str), format='%b-%Y', errors='coerce')
df['issue_year'] = df['issue_d'].dt.year
df = df[df['issue_year'] <= 2016]
log_step(4, 'เก็บเฉพาะสินเชื่อที่ออกปี 2007–2016', df)

df['term_months'] = pd.to_numeric(df['term'].astype(str).str.extract(r'(\d+)', expand=False), errors='coerce')
EMP_MAP = {'< 1 year': 0, '1 year': 1, '10+ years': 10}
EMP_MAP.update({f'{i} years': i for i in range(2, 10)})
df['emp_length_yrs'] = df['emp_length'].astype(str).str.strip().map(EMP_MAP)
assert df['term_months'].notna().any(), 'term_months ว่างทั้งหมด — ตรวจสอบคอลัมน์ term'
log_step(5, 'แปลงชนิดข้อมูล (วันที่ / term / emp_length)', df)

df = df.reset_index(drop=True)
display(pd.DataFrame(clean_log))
print(f'ข้อมูลที่ใช้: {len(df):,} แถว | อัตราผิดนัด = {df["default"].mean():.2%}')
TARGET = 'default'

# หลีกเลี่ยงการ del raw ทันที เผื่อต้องรันเซลล์นี้ซ้ำ
# del raw, issue_dt, ecl_dt; gc.collect();
gc.collect();
```

```python
# Descriptive statistics
NUM_EDA = ['loan_amnt', 'int_rate', 'installment', 'annual_inc', 'dti', 'fico_range_low',
           'delinq_2yrs', 'inq_last_6mths', 'open_acc', 'pub_rec', 'revol_bal', 'revol_util',
           'total_acc', 'mort_acc', 'pub_rec_bankruptcies', 'term_months', 'emp_length_yrs']
display(df[NUM_EDA].describe().T)

eda = df.sample(min(300_000, len(df)), random_state=RANDOM_STATE)   # ตัวอย่างสำหรับวาดกราฟ

# Histogram ของตัวแปรเชิงตัวเลข
hist_cols = ['loan_amnt', 'int_rate', 'annual_inc', 'dti', 'fico_range_low', 'revol_util', 'revol_bal', 'inq_last_6mths']
fig, axes = plt.subplots(2, 4, figsize=(20, 8))
for ax, c in zip(axes.ravel(), hist_cols):
    vals = eda[c].dropna()
    if c in ('annual_inc', 'revol_bal'):
        vals = np.log10(vals.clip(lower=0) + 1); ax.set_xlabel(f'log10({c}+1)')
    if c == 'dti':
        vals = vals[(vals >= 0) & (vals <= 100)]; ax.set_xlabel('dti (0–100)')
    ax.hist(vals, bins=50, color='#4C72B0'); ax.set_title(c)
plt.suptitle('Univariate distributions (histogram)', y=1.02); plt.tight_layout(); plt.show()

# Boxplot ของตัวแปรเชิงตัวเลข (ดู outlier)
box_cols = ['loan_amnt', 'int_rate', 'annual_inc', 'dti', 'revol_bal', 'revol_util', 'open_acc', 'total_acc']
fig, axes = plt.subplots(2, 4, figsize=(20, 7))
for ax, c in zip(axes.ravel(), box_cols):
    ax.boxplot(eda[c].dropna(), vert=False, flierprops={'markersize': 2, 'alpha': 0.3})
    ax.set_title(c); ax.set_yticks([])
plt.suptitle('Univariate boxplots (จุด = outlier ตามเกณฑ์ 1.5×IQR)', y=1.02); plt.tight_layout(); plt.show()

# ตาราง outlier: เกณฑ์ IQR 1.5 เท่า และ |Z| > 3
num_cols_eda = [c for c in df.select_dtypes('number').columns
                if c not in ('total_pymnt', 'default', 'issue_year')]
q1, q3 = df[num_cols_eda].quantile(0.25), df[num_cols_eda].quantile(0.75)
iqr = q3 - q1
z = (df[num_cols_eda] - df[num_cols_eda].mean()) / df[num_cols_eda].std()
out_tbl = pd.DataFrame({
    'min': df[num_cols_eda].min(), 'p50': df[num_cols_eda].median(),
    'p99': df[num_cols_eda].quantile(0.99), 'max': df[num_cols_eda].max(),
    'n_outlier_IQR': ((df[num_cols_eda] < q1 - 1.5*iqr) | (df[num_cols_eda] > q3 + 1.5*iqr)).sum(),
    'n_outlier_Z3': (z.abs() > 3).sum(),
})
out_tbl['pct_outlier_IQR'] = out_tbl['n_outlier_IQR'] / len(df) * 100
display(out_tbl.sort_values('pct_outlier_IQR', ascending=False))
print('Skewness:'); display(df[['annual_inc', 'revol_bal', 'loan_amnt', 'dti']].skew().to_frame('skew'))
del z

# ตัวแปรเชิงหมวดหมู่: ความถี่และความไม่สอดคล้องของหมวดหมู่
for c in ['term', 'home_ownership', 'emp_length', 'verification_status', 'application_type', 'grade']:
    display(df[c].value_counts(dropna=False).to_frame(f'n ({c})'))
```

## 4. Data Preprocessing
> **หลักการป้องกัน data leakage:** ขั้นตอนที่เป็น **กฎตายตัวระดับแถว** (ไม่เรียนรู้สถิติจากข้อมูล) ทำกับข้อมูลทั้งหมดได้ทันทีในหัวข้อนี้ ส่วนขั้นตอนที่ **ต้องเรียนรู้ค่าจากข้อมูล** (median, ขอบเขต IQR, หมวดหมู่ของ One-Hot, mean/SD ของ scaler) จะ **นิยามไว้เป็น Pipeline ในหัวข้อนี้ แต่ fit จาก train set เท่านั้น** หลังแบ่งข้อมูลในหัวข้อ 6

| ส่วน | วิธี | เรียนรู้จากข้อมูล? | เหตุผล |
|---|---|---|---|
| **ค่าที่เป็นไปไม่ได้** | `annual_inc` ≤ 0, `dti` < 0 หรือ > 100, `revol_util` > 150%, `earliest_cr_line` หลัง `issue_d` → ตั้งเป็นค่าว่าง (NaN) | ไม่ (กฎตายตัว) → ทำเลย | ขัดกับนิยามตัวแปร เป็นข้อมูลผิดพลาดเฉพาะคอลัมน์ ข้อมูลอื่นของแถวยังใช้ได้ จึงไม่ลบแถว แต่ให้ imputer จัดการ |
| **Missing values (ตัวเลข)** | `SimpleImputer(median, add_indicator=True)` | ใช่ → fit จาก train | median ทนต่อข้อมูลเบ้; indicator เก็บข้อมูลว่า "ไม่มีข้อมูล" |
| **Missing values (หมวดหมู่)** | `SimpleImputer(constant='MISSING')` | ใช่ → fit จาก train | ให้ค่าว่างเป็นหมวดหนึ่ง |
| **Outliers** | **IQR capping**: ตัดค่าที่เกิน [Q1 − 3·IQR, Q3 + 3·IQR] ให้เท่าขอบเขต (ข้ามตัวแปรที่ IQR = 0 เช่น จำนวนครั้งค้างชำระ) | ใช่ → fit จาก train | outlier ส่วนใหญ่เป็นค่าจริง (คนรายได้สูงมีจริง) จึงไม่ลบแถว แต่จำกัดอิทธิพลต่อโมเดลเชิงเส้น/ระยะทาง; ใช้ 3·IQR (ไม่ใช่ 1.5) เพราะข้อมูลการเงินเบ้ตามธรรมชาติ ร่วมกับ log transform ในหัวข้อ 5 |
| **Encoding** | `home_ownership`: ANY/NONE → OTHER (กฎตายตัว); `sub_grade` → ตัวเลขเชิงลำดับ (Label/Ordinal, หัวข้อ 5); ตัวแปรหมวดหมู่อื่น → `OneHotEncoder(handle_unknown='infrequent_if_exist', min_frequency=0.001)` | OHE: ใช่ → fit จาก train | หมวดที่น้อยมากรวมเป็น "infrequent", หมวดใหม่ใน test ไม่ทำให้ error |
| **Scaling** | `StandardScaler` | ใช่ → fit จาก train | จำเป็นสำหรับโมเดลที่ใช้ระยะทาง/สัมประสิทธิ์ (KNN, Logistic Regression, LDA, SMOTE) |

ขั้นที่ "เรียนรู้จากข้อมูล" ทั้งหมดห่อใน `Pipeline` / `ColumnTransformer` เดียวกับโมเดล → ใน cross-validation ตัว imputer / capper / scaler / encoder จะ **fit ใหม่ใน training fold ของแต่ละ fold** และ transform validation/test ด้วยค่าที่เรียนจาก train เท่านั้น

```python
# ============================================================
# 4.1 กฎตายตัวระดับแถว (Row-wise Data Cleaning)
# ค่าที่เป็นไปไม่ได้ -> NaN และรวมหมวดหมู่
# ============================================================

def fix_invalid(d):
    d = d.copy()

    rules = {
        'annual_inc': d['annual_inc'] <= 0,
        'dti': (d['dti'] < 0) | (d['dti'] > 100),
        'revol_util': d['revol_util'] > 150,
        'earliest_cr_line': d['earliest_cr_line'] > d['issue_d'],
    }

    counts = {}

    for col, mask in rules.items():
        counts[col] = int(mask.sum())
        d.loc[mask, col] = np.nan

    d['home_ownership'] = (
        d['home_ownership']
        .astype(str)
        .replace({'ANY': 'OTHER', 'NONE': 'OTHER'})
    )

    return d, counts


df, inv_counts = fix_invalid(df)

print('จำนวนค่าที่เป็นไปไม่ได้ที่ถูกตั้งเป็น NaN (ไม่ลบแถว):')
display(pd.Series(inv_counts, name='n').to_frame())

display(
    df['home_ownership']
    .value_counts()
    .to_frame('n (home_ownership หลังรวมหมวด)')
)


# ============================================================
# 4.2 IQR Capping
# เรียนรู้ขอบเขตจากข้อมูลที่ใช้ fit เท่านั้น
# ============================================================

class IQRCapper(BaseEstimator, TransformerMixin):

    def __init__(self, k=3.0):
        self.k = k

    def fit(self, X, y=None):
        X = np.asarray(X, dtype=float)

        q1 = np.nanpercentile(X, 25, axis=0)
        q3 = np.nanpercentile(X, 75, axis=0)

        iqr = q3 - q1

        self.lower_ = np.where(
            iqr > 0,
            q1 - self.k * iqr,
            -np.inf
        )

        self.upper_ = np.where(
            iqr > 0,
            q3 + self.k * iqr,
            np.inf
        )

        return self

    def transform(self, X):
        X = np.asarray(X, dtype=float)
        return np.clip(X, self.lower_, self.upper_)

    def get_feature_names_out(self, input_features=None):
        return np.asarray(input_features, dtype=object)


# ============================================================
# 4.3 Preprocessing Pipeline
# ตัวเลข: Impute -> IQR Capping -> Scaling
# หมวดหมู่: Impute -> Label Encoding
# ============================================================

def make_preprocess(num_cols, cat_cols):

    # ตัวแปรตัวเลข
    num_pipe = Pipeline([
        ('impute', SimpleImputer(
            strategy='median',
            add_indicator=True
        )),
        ('cap', IQRCapper(k=3.0)),
        ('scale', StandardScaler())
    ])

    # ตัวแปรหมวดหมู่: Label Encoding
    cat_pipe = Pipeline([
        ('impute', SimpleImputer(
            strategy='constant',
            fill_value='MISSING'
        )),
        ('label', OrdinalEncoder(
            handle_unknown='use_encoded_value',
            unknown_value=-1
        ))
    ])

    return ColumnTransformer([
        ('num', num_pipe, num_cols),
        ('cat', cat_pipe, cat_cols)
    ])


print('นิยาม make_preprocess() เรียบร้อย')
print('จะ fit จาก Train set ในหัวข้อ 6.3')
```

### 5.2 Feature Selection
1. **Drop high correlation:** หาคู่ feature ที่ |Spearman corr| > 0.9 แล้วตัดตัวที่สัมพันธ์กับ target น้อยกว่าออก (ลด multicollinearity และความซ้ำซ้อน)
2. **Feature Importance:** เทรน HistGradientBoosting บน 70% ของตัวอย่าง train แล้วคำนวณ **permutation importance** บนอีก 30% ของ train → ตัด feature ที่ importance ≤ 0 (ไม่ช่วยให้ ROC-AUC ดีขึ้นเลย)

> ⚠️ ทั้ง 2 วิธี **ใช้ target ในการตัดสินใจ** ถ้าคำนวณจากข้อมูลทั้งหมดจะเกิด data leakage (test มีส่วนเลือก feature) — จึงนิยามขั้นตอนไว้ที่นี่ และ **รันบน train set เท่านั้นในหัวข้อ 6.2** หลังแบ่งข้อมูล

```python
# 5.1 Feature extraction (row-wise)
def add_features(d):
    d = d.copy()
    sg = d['sub_grade'].astype(str)
    d['sub_grade_num'] = (sg.str[0].map({g: i for i, g in enumerate('ABCDEFG')}) * 5
                          + pd.to_numeric(sg.str[1], errors='coerce'))
    d['issue_month'] = d['issue_d'].dt.month
    d['fico_avg'] = (d['fico_range_low'] + d['fico_range_high']) / 2
    d['credit_hist_yrs'] = (d['issue_d'] - d['earliest_cr_line']).dt.days / 365.25
    d['log_annual_inc'] = np.log1p(d['annual_inc'])
    d['log_revol_bal'] = np.log1p(d['revol_bal'])
    d['loan_to_income'] = d['loan_amnt'] / d['annual_inc']
    d['installment_to_income'] = d['installment'] / (d['annual_inc'] / 12)
    d['revol_bal_to_income'] = d['revol_bal'] / d['annual_inc']
    d['open_acc_ratio'] = d['open_acc'] / d['total_acc'].replace(0, np.nan)
    return d

df = add_features(df)

NUM_FEATURES = ['loan_amnt', 'term_months', 'int_rate', 'installment', 'sub_grade_num', 'issue_month',
                'emp_length_yrs', 'log_annual_inc', 'dti', 'delinq_2yrs', 'fico_avg',
                'inq_last_6mths', 'open_acc', 'pub_rec', 'log_revol_bal', 'revol_util',
                'total_acc', 'mort_acc', 'pub_rec_bankruptcies', 'credit_hist_yrs',
                'loan_to_income', 'installment_to_income', 'revol_bal_to_income', 'open_acc_ratio']
CAT_FEATURES = ['home_ownership', 'verification_status', 'purpose', 'addr_state',
                'application_type', 'initial_list_status']
for c in CAT_FEATURES:
    df[c] = df[c].astype(str).replace({'nan': np.nan}).astype(object)
print(f'จำนวน feature เริ่มต้น: {len(NUM_FEATURES)} ตัวเลข + {len(CAT_FEATURES)} หมวดหมู่')

new_feats = ['sub_grade_num', 'issue_month', 'fico_avg', 'credit_hist_yrs', 'log_annual_inc', 'loan_to_income',
             'installment_to_income', 'revol_bal_to_income', 'open_acc_ratio']
display(df[new_feats].describe().T)
display(df.groupby('issue_month')[TARGET].mean().to_frame('default rate').T)

# 5.2 Feature selection (ฟังก์ชัน) — จะเรียกใช้กับ train set ในหัวข้อ 6.2
def select_features(train_part, num_features, cat_features, corr_thr=0.9, n_sample=200_000):
    fs_sample = train_part.sample(min(n_sample, len(train_part)), random_state=RANDOM_STATE)
    # (1) drop high correlation
    corr_tr = fs_sample[num_features].corr(method='spearman').abs()
    target_corr = fs_sample[num_features].corrwith(fs_sample[TARGET], method='spearman').abs()
    drop_corr, pairs = set(), []
    for i in range(len(num_features)):
        for j in range(i + 1, len(num_features)):
            a, b = num_features[i], num_features[j]
            if corr_tr.loc[a, b] > corr_thr and a not in drop_corr and b not in drop_corr:
                drop = a if target_corr[a] < target_corr[b] else b
                drop_corr.add(drop)
                pairs.append({'feature_1': a, 'feature_2': b, '|corr|': corr_tr.loc[a, b], 'ตัดออก': drop})
    num_after = [c for c in num_features if c not in drop_corr]
    # (2) permutation importance (HGB บน 70% ของตัวอย่าง train, วัดบนอีก 30% ของ train)
    fi_tr, fi_ho = train_test_split(fs_sample, test_size=0.3, stratify=fs_sample[TARGET], random_state=RANDOM_STATE)
    cand = num_after + cat_features
    fi_pipe = Pipeline([('prep', make_preprocess(num_after, cat_features)),
                        ('model', HistGradientBoostingClassifier(random_state=RANDOM_STATE))]).fit(fi_tr[cand], fi_tr[TARGET])
    fi = permutation_importance(fi_pipe, fi_ho[cand], fi_ho[TARGET], scoring='roc_auc', n_repeats=3,
                                random_state=RANDOM_STATE, n_jobs=1)
    fi_df = pd.DataFrame({'feature': cand, 'importance (ΔAUC)': fi.importances_mean, 'sd': fi.importances_std}) \
              .sort_values('importance (ΔAUC)', ascending=False).set_index('feature')
    drop_imp = fi_df.index[fi_df['importance (ΔAUC)'] <= 0].tolist()
    fi_df['ผล'] = np.where(fi_df.index.isin(drop_imp), 'ตัดออก (importance ≤ 0)', 'เก็บไว้')
    num_sel = [c for c in num_after if c not in drop_imp]
    cat_sel = [c for c in cat_features if c not in drop_imp]
    return num_sel, cat_sel, pd.DataFrame(pairs), fi_df, sorted(drop_corr), drop_imp
```

## 7. Handling Imbalanced Data
> **ทำเฉพาะกับ Training Set เท่านั้น!** — validation และ test ต้องคงสัดส่วนจริง (~20% ผิดนัด) เพื่อให้ผลประเมินสะท้อนการใช้งานจริง

| วิธี | ระดับ | รายละเอียด | ข้อสังเกต |
|---|---|---|---|
| ไม่ปรับ (`none`) | – | เทรนบนสัดส่วนจริง แล้วชดเชยด้วยการเลือก threshold | ใช้เป็นตัวเทียบ |
| **Oversampling: SMOTE** (`smote`) | Data level | สร้างข้อมูลสังเคราะห์ของคลาสส่วนน้อย (ผิดนัด) โดย interpolate ระหว่างเพื่อนบ้านใกล้ที่สุด จนคลาสเท่ากัน | ต้องทำ **หลัง** encoding/scaling (ใช้ระยะทาง) และทำ **ภายใน training set เท่านั้น**; ตัวแปร one-hot จะได้ค่าทศนิยม (SMOTENC เป็นทางเลือก) — ADASYN คล้ายกันแต่เน้นจุดที่จำแนกยาก จึงเลือกทดลองแค่ SMOTE |
| **Undersampling** (`undersample`) | Data level | สุ่มตัดคลาสส่วนใหญ่ (ชำระครบ) ให้เท่ากับคลาสส่วนน้อย | เร็ว แต่ทิ้งข้อมูลไปมาก — ข้อมูลเรามีมากพอ |
| **`class_weight='balanced'`** (`class_weight`) | Model level | ถ่วงน้ำหนัก loss ของคลาสส่วนน้อยให้มากขึ้น (XGBoost ใช้ `scale_pos_weight`) | ไม่เปลี่ยนข้อมูล ใช้ได้กับโมเดลที่รองรับ |

ในหัวข้อนี้เตรียมฟังก์ชันทั้ง 3 วิธี — **การเลือกวิธีที่ดีที่สุดทำในหัวข้อ 9.2** ด้วย cross-validation บน train (resample เฉพาะ training fold)

⚠️ **สำคัญต่อคำถามของเรา:** class_weight / undersampling / SMOTE ทำให้โมเดล "เห็น" ผู้ผิดนัด ~50% ระหว่างเทรน ทั้งที่จริงมี ~20% → **% ผิดนัดที่ทำนายจะสูงเกินจริง** จึงต้อง **calibrate** ความน่าจะเป็นด้วย validation set (หัวข้อ 10.2) ก่อนรายงานเป็น % ผิดนัดของลูกค้า

```python
y_train_all = train_df[TARGET].values

# (1) Undersampling — ใช้กับ train set เท่านั้น
def undersample(X, y, random_state=RANDOM_STATE):
    """สุ่มลดคลาส 0 ให้เท่ากับคลาส 1"""
    rng = np.random.RandomState(random_state)
    pos = np.where(y == 1)[0]; neg = np.where(y == 0)[0]
    idx = np.concatenate([pos, rng.choice(neg, size=len(pos), replace=False)])
    rng.shuffle(idx)
    return X.iloc[idx], y[idx]

# (2) SMOTE — ใช้ imbalanced-learn ถ้ามี, ถ้าไม่มีใช้ฉบับที่เขียนเอง (หลักการเดียวกัน)
class SimpleSMOTE:
    """SMOTE แบบย่อ: สุ่มจุดคลาสส่วนน้อย + เพื่อนบ้าน k ตัว แล้วสร้างจุดใหม่บนเส้นที่เชื่อมกัน"""
    def __init__(self, k_neighbors=5, random_state=RANDOM_STATE):
        self.k_neighbors, self.random_state = k_neighbors, random_state
    def fit_resample(self, X, y):
        rng = np.random.RandomState(self.random_state)
        X = np.asarray(X, dtype=float); y = np.asarray(y)
        X_min = X[y == 1]; n_new = int((y == 0).sum() - (y == 1).sum())
        if n_new <= 0: return X, y
        nn = NearestNeighbors(n_neighbors=self.k_neighbors + 1).fit(X_min)
        nbrs = nn.kneighbors(X_min, return_distance=False)[:, 1:]
        base = rng.randint(0, len(X_min), n_new)
        nb = nbrs[base, rng.randint(0, self.k_neighbors, n_new)]
        gap = rng.rand(n_new, 1)
        X_new = X_min[base] + gap * (X_min[nb] - X_min[base])
        return np.vstack([X, X_new]), np.concatenate([y, np.ones(n_new, dtype=y.dtype)])

def make_smote():
    return SMOTE(random_state=RANDOM_STATE) if HAS_IMB else SimpleSMOTE()

# (3) class_weight — คำนวณจาก train
NEG_POS_RATIO = (y_train_all == 0).sum() / (y_train_all == 1).sum()
cw = len(y_train_all) / (2 * np.bincount(y_train_all))

# ฟังก์ชันรวม: fit pipeline (preprocess + model) ด้วยวิธีจัดการ imbalance ที่เลือก — เรียกกับข้อมูล train เท่านั้น
IMB_STRATEGIES = ['none', 'class_weight', 'undersample', 'smote']
def fit_with_strategy(pipe, X, y, strategy):
    m = clone(pipe)
    if strategy == 'none':
        return m.fit(X, y)
    if strategy == 'class_weight':
        params = m.named_steps['model'].get_params()
        if 'class_weight' in params:       m.set_params(model__class_weight='balanced')
        elif 'scale_pos_weight' in params: m.set_params(model__scale_pos_weight=NEG_POS_RATIO)
        else: return None                  # โมเดลนี้ไม่รองรับการถ่วงน้ำหนักคลาส
        return m.fit(X, y)
    if strategy == 'undersample':
        Xu, yu = undersample(X, y)
        return m.fit(Xu, yu)
    if strategy == 'smote':                # preprocess (fit บน train) -> SMOTE -> model
        prep = m.named_steps['prep'].fit(X)
        Zs, ys = make_smote().fit_resample(prep.transform(X), y)
        model = m.named_steps['model'].fit(Zs, ys)
        return Pipeline([('prep', prep), ('model', model)])
    raise ValueError(strategy)

# สรุปผลของแต่ละวิธีต่อจำนวนแถวใน train set
n0, n1 = (y_train_all == 0).sum(), (y_train_all == 1).sum()
display(pd.DataFrame({
    'วิธี': ['none (ข้อมูลจริง)', 'undersample', 'smote', 'class_weight=balanced'],
    'คลาส 0 (Fully Paid)': [n0, n1, n0, f'{n0:,} (น้ำหนัก {cw[0]:.3f})'],
    'คลาส 1 (Charged Off)': [n1, n1, n0, f'{n1:,} (น้ำหนัก {cw[1]:.3f})'],
    'รวม (แถวที่ใช้เทรน)': [n0 + n1, 2 * n1, 2 * n0, n0 + n1],
}).set_index('วิธี'))
print(f'neg/pos ratio ใน train = {NEG_POS_RATIO:.2f} | SMOTE ใช้ {"imbalanced-learn" if HAS_IMB else "SimpleSMOTE (เขียนเอง)"}')
print('validation / test ไม่ถูก resample — คงสัดส่วนจริง:', f'{val_df[TARGET].mean():.2%}', '/', f'{test_df[TARGET].mean():.2%}')
```

## 8. Model Selection
รัน **9 อัลกอริทึม** ด้วย **default parameters** + baseline (Dummy) ด้วย **Stratified 3-fold CV** บนตัวอย่าง 60,000 แถวจาก train แล้ว **เปรียบเทียบด้วยตัวชี้วัดประสิทธิภาพหลายตัว** เพื่อเลือก **3 โมเดลที่ดีที่สุด** ไป tune ต่อ (ขั้นนี้ยังไม่ปรับ imbalance — ทุกโมเดลอยู่ในเงื่อนไขเดียวกัน จึงเทียบกันได้อย่างยุติธรรม)

| # | โมเดล | เหตุผลที่เลือกมาเปรียบเทียบ |
|---|---|---|
| 0 | Dummy (prior) | baseline: เดาตามสัดส่วนคลาส |
| 1 | Logistic Regression | มาตรฐานของ credit scoring ตีความได้ง่าย |
| 2 | Linear Discriminant Analysis | โมเดลเชิงเส้นแบบ generative เร็วมาก |
| 3 | Gaussian Naive Bayes | baseline เชิงความน่าจะเป็นที่ง่ายที่สุด |
| 4 | K-Nearest Neighbors | non-parametric อิงระยะทาง |
| 5 | Decision Tree | ตีความเป็นกฎได้ จับความสัมพันธ์ไม่เชิงเส้น |
| 6 | Random Forest | ensemble แบบ bagging ลด variance ของต้นไม้ |
| 7 | Extra Trees | bagging แบบสุ่มจุดแบ่ง เร็วกว่า RF |
| 8 | HistGradientBoosting | boosting แบบ histogram ออกแบบสำหรับข้อมูลใหญ่ |
| 9 | XGBoost | boosting ที่นิยมที่สุดสำหรับข้อมูลตาราง (ถ้าติดตั้งอยู่ — Colab มีให้แล้ว) |

**ตัวชี้วัดที่ใช้เปรียบเทียบ (ทุกตัวคำนวณด้วย 3-fold CV)**

| กลุ่ม | ตัวชี้วัด | ดีเมื่อ | วัดอะไร / ทำไมใช้ |
|---|---|---|---|
| จัดลำดับความเสี่ยง | **ROC-AUC** | สูง | แยกคนผิดนัดออกจากคนชำระครบได้ดีแค่ไหน ไม่ขึ้นกับ threshold (ตัวชี้วัดหลัก) |
| | **PR-AUC** | สูง | เน้นคลาสผิดนัดซึ่งเป็นคลาสส่วนน้อย (ค่าเดาสุ่ม = อัตราผิดนัด ~0.20) |
| ความแม่นของ % | **Brier score** | ต่ำ | ค่าเฉลี่ยของ (ความน่าจะเป็นที่ทำนาย − ผลจริง)² — ตรงกับคำถาม "ผิดนัดกี่ %" |
| | **Log loss** | ต่ำ | ลงโทษหนักเมื่อมั่นใจผิด เช่น ทาย 1% แต่ผิดนัดจริง |
| การตัดสินใจ (ที่ threshold = อัตราผิดนัดใน train) | **Recall** | สูง | จับคนที่ผิดนัดจริงได้กี่ % (สำคัญเพราะปล่อยกู้ให้คนผิดนัด = เสียเงินต้น) |
| | **Precision** | สูง | ในคนที่ถูกทายว่าเสี่ยง ผิดนัดจริงกี่ % |
| | **F1-Score** | สูง | ค่าเฉลี่ยฮาร์มอนิกของ Precision กับ Recall (ต้องดูคู่กันเสมอเมื่อข้อมูลไม่สมดุล) |
| | **Balanced Accuracy** | สูง | ค่าเฉลี่ยของ recall ทั้ง 2 คลาส ใช้แทน Accuracy ที่ถูกหลอกโดย imbalance |
| ความเสถียร / ต้นทุน | SD ของ ROC-AUC, ช่องว่าง train − CV, เวลาที่ใช้ | ต่ำ | ดู overfitting และความคุ้มค่าในการใช้งานจริง |

* threshold-dependent metrics (Recall/Precision/F1/Balanced Acc) ใช้ threshold = **อัตราผิดนัดใน train (~0.20)** แทน 0.5 เพราะโมเดลที่ไม่ได้ปรับ imbalance แทบไม่เคยให้ความน่าจะเป็นเกิน 0.5 ถ้าใช้ 0.5 ทุกโมเดลจะได้ recall ใกล้ 0 และเทียบกันไม่ได้
* **วิธีเลือก:** จัดอันดับแต่ละโมเดลในทุกตัวชี้วัด (8 ตัว) แล้วเลือก **3 โมเดลที่มีอันดับเฉลี่ยดีที่สุด** (ถ้าเสมอกันดู ROC-AUC) — เพื่อไม่ให้ตัดสินจากตัวชี้วัดตัวเดียว

```python
# 8. Model selection: เปรียบเทียบหลายอัลกอริทึม (default parameters) ด้วยตัวชี้วัดหลายตัว
from sklearn.metrics import make_scorer, balanced_accuracy_score
SCREEN_N, TUNE_N = 60_000, 150_000
X_tune, _, y_tune, _ = train_test_split(X_train, y_train, train_size=min(TUNE_N, len(y_train) - 2),
                                        stratify=y_train, random_state=RANDOM_STATE)
X_screen, _, y_screen, _ = train_test_split(X_tune, y_tune, train_size=min(SCREEN_N, len(y_tune) - 2),
                                            stratify=y_tune, random_state=RANDOM_STATE)
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)

PREV = float(y_train.mean())          # threshold สำหรับ metric ที่ต้องตัดสินใจ = อัตราผิดนัดใน train
def _at_thr(metric):
    def f(y, p): return metric(y, (np.asarray(p) >= PREV).astype(int))
    return f
SCORING = {
    'roc_auc': 'roc_auc', 'pr_auc': 'average_precision',
    'brier': 'neg_brier_score', 'log_loss': 'neg_log_loss',
    'recall': make_scorer(_at_thr(recall_score), response_method='predict_proba'),
    'precision': make_scorer(_at_thr(lambda y, yp: precision_score(y, yp, zero_division=0)), response_method='predict_proba'),
    'f1': make_scorer(_at_thr(f1_score), response_method='predict_proba'),
    'bal_acc': make_scorer(_at_thr(balanced_accuracy_score), response_method='predict_proba'),
}
METRIC_COLS = {'roc_auc': 'ROC-AUC', 'pr_auc': 'PR-AUC', 'brier': 'Brier', 'log_loss': 'Log loss',
               'recall': 'Recall', 'precision': 'Precision', 'f1': 'F1', 'bal_acc': 'Balanced Acc'}
LOWER_BETTER = ['Brier', 'Log loss']

screen_models = {
    'Dummy (prior)': DummyClassifier(strategy='prior'),
    'LogisticRegression': LogisticRegression(max_iter=2000),
    'LDA': LinearDiscriminantAnalysis(),
    'GaussianNB': GaussianNB(),
    'KNN': KNeighborsClassifier(n_jobs=-1),
    'DecisionTree': DecisionTreeClassifier(random_state=RANDOM_STATE),
    'RandomForest': RandomForestClassifier(n_jobs=-1, random_state=RANDOM_STATE),
    'ExtraTrees': ExtraTreesClassifier(n_jobs=-1, random_state=RANDOM_STATE),
    'HistGradientBoosting': HistGradientBoostingClassifier(random_state=RANDOM_STATE),
}
if HAS_XGB:
    screen_models['XGBoost'] = XGBClassifier(n_jobs=-1, random_state=RANDOM_STATE, tree_method='hist',
                                             eval_metric='logloss')

screen_rows, screen_fitted = [], {}
for name, est in screen_models.items():
    pipe = Pipeline([('prep', clone(preprocess)), ('model', est)])
    t0 = time.time()
    r = cross_validate(pipe, X_screen, y_screen, cv=cv, scoring=SCORING, return_train_score=True, n_jobs=1)
    screen_fitted[name] = clone(pipe).fit(X_screen, y_screen)     # ใช้เปรียบเทียบใน Diagnosis (หัวข้อ 12)
    row = {'model': name}
    for k, lab in METRIC_COLS.items():
        v = r[f'test_{k}'].mean()
        row[lab] = -v if lab in LOWER_BETTER else v               # sklearn คืนค่าติดลบสำหรับ loss
    row.update({'CV ROC-AUC SD': r['test_roc_auc'].std(), 'train ROC-AUC': r['train_roc_auc'].mean(),
                'fold ROC-AUC': [round(float(v), 4) for v in r['test_roc_auc']], 'time (s)': round(time.time() - t0, 1)})
    row['train − CV gap'] = row['train ROC-AUC'] - row['ROC-AUC']
    screen_rows.append(row)
    print(f'{name:22s} ROC-AUC {row["ROC-AUC"]:.4f} | PR-AUC {row["PR-AUC"]:.4f} | Brier {row["Brier"]:.4f} | '
          f'F1 {row["F1"]:.4f} | Recall {row["Recall"]:.4f} | {row["time (s)"]}s')

screen_df = pd.DataFrame(screen_rows).set_index('model')
screen_df['CV ROC-AUC mean'] = screen_df['ROC-AUC']              # ชื่อคอลัมน์ที่หัวข้อ 12 ใช้

# จัดอันดับในทุกตัวชี้วัด แล้วหาอันดับเฉลี่ย (ไม่นับ Dummy ซึ่งเป็น baseline)
cand_df = screen_df.drop(index='Dummy (prior)')
ranks = pd.DataFrame({lab: cand_df[lab].rank(ascending=(lab in LOWER_BETTER), method='min')
                      for lab in METRIC_COLS.values()})
cand_df = cand_df.assign(**{'อันดับเฉลี่ย': ranks.mean(axis=1)})
cand_df = cand_df.sort_values(['อันดับเฉลี่ย', 'ROC-AUC'], ascending=[True, False])
show_cols = list(METRIC_COLS.values()) + ['CV ROC-AUC SD', 'train ROC-AUC', 'train − CV gap', 'time (s)', 'อันดับเฉลี่ย']
display(pd.concat([cand_df[show_cols], screen_df.loc[['Dummy (prior)'], show_cols[:-1]]]).round(4))
display(ranks.loc[cand_df.index].astype(int).assign(**{'อันดับเฉลี่ย': cand_df['อันดับเฉลี่ย']})
        .rename_axis('อันดับในแต่ละตัวชี้วัด (1 = ดีที่สุด)'))

# กราฟ: heatmap ของตัวชี้วัด (สีตามอันดับ, ตัวเลข = ค่าจริง)
fig, ax = plt.subplots(1, 2, figsize=(20, 6), gridspec_kw={'width_ratios': [3, 2]})
sns.heatmap(ranks.loc[cand_df.index], annot=cand_df[list(METRIC_COLS.values())].round(3), fmt='',
            cmap='RdYlGn_r', cbar_kws={'label': 'rank (1 = best)'}, ax=ax[0])
ax[0].set_title('Model selection: CV metrics (color = rank, text = value)'); ax[0].set_ylabel('')
plot_df = screen_df.loc[list(cand_df.index[::-1]) + ['Dummy (prior)']]
ax[1].barh(plot_df.index, plot_df['ROC-AUC'], xerr=plot_df['CV ROC-AUC SD'], color='#4C72B0', label='CV ROC-AUC')
ax[1].scatter(plot_df['train ROC-AUC'], plot_df.index, color='#DD8452', zorder=3, label='train ROC-AUC')
ax[1].axvline(0.5, ls='--', c='k'); ax[1].set_xlim(0.4, 1.02); ax[1].legend(loc='lower right')
ax[1].set_title('Train vs CV ROC-AUC (gap = overfitting)')
plt.tight_layout(); plt.show()

TOP3 = cand_df.index[:3].tolist()
print('3 โมเดลที่ดีที่สุด (อันดับเฉลี่ยจาก 8 ตัวชี้วัด):', TOP3)
```

**การวิเคราะห์ผล model selection** — ดูช่องว่างระหว่าง train กับ CV: โมเดลที่ train ROC-AUC ≈ 1.0 แต่ CV ต่ำกว่ามาก (เช่น Decision Tree, Random Forest, Extra Trees, KNN แบบ default ที่ไม่จำกัดความลึก/เพื่อนบ้านน้อย) คือ **overfitting** ส่วนโมเดลที่ทั้ง train และ CV ต่ำ (เช่น Naive Bayes) คือ **underfitting** — ขั้น Hyperparameter Tuning (หัวข้อ 9) จะควบคุมความซับซ้อน (เช่น `max_depth`, `min_samples_leaf`, `n_neighbors`)

## 9. Hyperparameter Tuning
### 9.1 Tuning ด้วย Grid / Random Search + Cross-Validation
* ใช้ **Cross-Validation (Stratified 3-fold)** บนตัวอย่าง 150,000 แถวจาก train ปรับจูนจนได้ **Best Params** ของ 3 โมเดลที่เลือกจากหัวข้อ 8 (เลือกด้วย CV ROC-AUC)
* **Search Strategy**
  * **Grid Search** — ลองทุกค่า ช้าแต่ได้ผลดีแน่นอน → ใช้เมื่อจำนวน combination ไม่มาก (≤ 8 เช่น Logistic Regression `C` 5 ค่า, LDA `shrinkage` 4 ค่า)
  * **Random Search** — สุ่มค่า (8 รอบ) เร็วและมักได้ผลดีพอ ๆ กัน → ใช้เมื่อ grid ใหญ่ (เช่น HistGradientBoosting 81 combinations) เพราะข้อมูลมีขนาดใหญ่ การลองทุกค่าใช้เวลานานเกินไปบน Colab
* ทุกการค้นหาใช้ `Pipeline(preprocess + model)` → preprocessing fit ใหม่ในแต่ละ fold (ไม่มี leakage)

```python
# 9. Hyperparameter tuning ของ 3 โมเดลที่ดีที่สุด (Grid ถ้า combination ≤ N_ITER, ไม่เช่นนั้น Random)
TUNE_BASE = {
    'LogisticRegression': LogisticRegression(max_iter=2000),
    'LDA': LinearDiscriminantAnalysis(solver='lsqr'),
    'GaussianNB': GaussianNB(),
    'KNN': KNeighborsClassifier(n_jobs=-1),
    'DecisionTree': DecisionTreeClassifier(random_state=RANDOM_STATE),
    'RandomForest': RandomForestClassifier(n_estimators=150, max_samples=0.5, n_jobs=-1, random_state=RANDOM_STATE),
    'ExtraTrees': ExtraTreesClassifier(n_estimators=150, n_jobs=-1, random_state=RANDOM_STATE),
    'HistGradientBoosting': HistGradientBoostingClassifier(max_iter=500, early_stopping=True, validation_fraction=0.1,
                                                           n_iter_no_change=20, random_state=RANDOM_STATE),
}
if HAS_XGB:
    TUNE_BASE['XGBoost'] = XGBClassifier(n_estimators=400, n_jobs=-1, random_state=RANDOM_STATE,
                                         tree_method='hist', eval_metric='logloss')
PARAM_GRIDS = {
    'LogisticRegression': {'model__C': [0.001, 0.01, 0.1, 1, 10]},
    'LDA': {'model__shrinkage': [None, 'auto', 0.1, 0.5]},
    'GaussianNB': {'model__var_smoothing': [1e-9, 1e-7, 1e-5, 1e-3, 1e-1]},
    'KNN': {'model__n_neighbors': [25, 50, 100, 200], 'model__weights': ['uniform', 'distance']},
    'DecisionTree': {'model__max_depth': [5, 8, 12, 16], 'model__min_samples_leaf': [50, 200, 500],
                     'model__criterion': ['gini', 'entropy']},
    'RandomForest': {'model__max_depth': [10, 14, 18], 'model__min_samples_leaf': [20, 50, 100],
                     'model__max_features': ['sqrt', 0.3]},
    'ExtraTrees': {'model__max_depth': [12, 16, 20], 'model__min_samples_leaf': [20, 50, 100],
                   'model__max_features': ['sqrt', 0.5]},
    'HistGradientBoosting': {'model__learning_rate': [0.03, 0.05, 0.1], 'model__max_leaf_nodes': [15, 31, 63],
                             'model__min_samples_leaf': [50, 200, 500], 'model__l2_regularization': [0.0, 1.0, 5.0]},
    'XGBoost': {'model__max_depth': [3, 5, 7], 'model__learning_rate': [0.03, 0.1], 'model__subsample': [0.7, 1.0],
                'model__colsample_bytree': [0.7, 1.0], 'model__min_child_weight': [1, 10, 50]},
}
FIT_CAP = {'KNN': 150_000}     # KNN ทำนายช้ามากเมื่อข้อมูล train ใหญ่ จึงจำกัดจำนวนแถว
N_ITER = 8

search_results, tuned_rows = {}, []
for name in TOP3:
    pipe = Pipeline([('prep', clone(preprocess)), ('model', clone(TUNE_BASE[name]))])
    grid = PARAM_GRIDS[name]
    n_comb = len(ParameterGrid(grid))
    if n_comb <= N_ITER:
        search = GridSearchCV(pipe, grid, scoring=SCORING, refit='roc_auc', cv=cv, n_jobs=1)
    else:
        search = RandomizedSearchCV(pipe, grid, n_iter=N_ITER, scoring=SCORING, refit='roc_auc', cv=cv,
                                    n_jobs=1, random_state=RANDOM_STATE)
    t0 = time.time(); search.fit(X_tune, y_tune); secs = time.time() - t0
    search_results[name] = search
    i, r = search.best_index_, search.cv_results_
    tuned_rows.append({'model': name, 'search': type(search).__name__,
                       'best_params': {k.replace('model__', ''): v for k, v in search.best_params_.items()},
                       'CV ROC-AUC mean': r['mean_test_roc_auc'][i], 'CV ROC-AUC SD': r['std_test_roc_auc'][i],
                       'CV PR-AUC mean': r['mean_test_pr_auc'][i],
                       'fold ROC-AUC': [round(float(r[f'split{k}_test_roc_auc'][i]), 4) for k in range(cv.n_splits)],
                       'tuning time (s)': round(secs)})
    print(f'{name}: tuned CV ROC-AUC = {r["mean_test_roc_auc"][i]:.4f} ± {r["std_test_roc_auc"][i]:.4f} ({secs:.0f}s)')
tuned_cv_df = pd.DataFrame(tuned_rows).set_index('model')
display(tuned_cv_df)
```

### 9.2 เลือกวิธีจัดการ Imbalanced Data ด้วย Cross-Validation (ใช้ train เท่านั้น)
วิธีจัดการ imbalance (หัวข้อ 7) ถือเป็น "hyperparameter" อีกตัวหนึ่ง จึงเลือกด้วย **Stratified 3-fold CV บนข้อมูล train ชุดเดียวกับที่ใช้ tune** สำหรับแต่ละโมเดลใน 3 อันดับแรก (ใช้ Best Params จาก 9.1)
* resampling (undersample / SMOTE) ทำ **เฉพาะใน training fold** — fold ที่ใช้วัดผลคงสัดส่วนจริง
* เกณฑ์เลือก: **CV PR-AUC** สูงสุด (เน้นคลาสผิดนัด); ถ้าต่างกันไม่ถึง 0.002 เลือกวิธีที่ง่ายกว่า (`none` → `class_weight` → `undersample` → `smote`) เพราะ % ผิดนัดจะถูก calibrate ภายหลังอยู่แล้ว

```python
# 9.2 CV ของวิธีจัดการ imbalance (resample เฉพาะ training fold)
def cv_strategy(pipe, X, y, strategy, cv=cv):
    aucs, aps = [], []
    for tr_i, te_i in cv.split(X, y):
        m = fit_with_strategy(pipe, X.iloc[tr_i], y[tr_i], strategy)
        if m is None: return None
        p = m.predict_proba(X.iloc[te_i])[:, 1]
        aucs.append(roc_auc_score(y[te_i], p)); aps.append(average_precision_score(y[te_i], p))
    return np.mean(aucs), np.std(aucs), np.mean(aps)

imb_rows, BEST_STRAT = [], {}
for name in TOP3:
    best_pipe = search_results[name].best_estimator_
    rows = []
    for strategy in IMB_STRATEGIES:
        t0 = time.time(); r = cv_strategy(best_pipe, X_tune, y_tune, strategy)
        if r is None: continue
        rows.append({'model': name, 'strategy': strategy, 'CV ROC-AUC': r[0], 'CV ROC-AUC SD': r[1],
                     'CV PR-AUC': r[2], 'time (s)': round(time.time() - t0)})
    sdf = pd.DataFrame(rows)
    best_ap = sdf['CV PR-AUC'].max()
    BEST_STRAT[name] = sdf.loc[sdf['CV PR-AUC'] >= best_ap - 0.002, 'strategy'].iloc[0]   # ง่ายที่สุดในกลุ่มที่ดีพอ ๆ กัน
    imb_rows += rows
imb_df = pd.DataFrame(imb_rows).set_index(['model', 'strategy'])
display(imb_df)
print('วิธีจัดการ imbalance ที่เลือก (จาก CV):', BEST_STRAT)
```

## 10. Model Retraining
นำ **Best Params** (หัวข้อ 9.1) + **วิธีจัดการ imbalance ที่เลือก** (หัวข้อ 9.2) มา **เทรนโมเดลใหม่ด้วย training set ทั้งหมด** (รวมทุก fold ที่เคยถูกแบ่งทำ CV และส่วนที่ไม่ได้อยู่ในตัวอย่าง tuning) — โมเดลจึงได้เรียนจากข้อมูลมากที่สุดก่อนนำไปใช้

จากนั้นใช้ **validation set** (ยังไม่แตะ test) เพื่อ
1. **10.1** เปรียบเทียบ 3 โมเดลที่ retrain แล้ว → เลือกโมเดลสุดท้าย
2. **10.2** Calibrate คะแนนของโมเดลให้เป็น **% ผิดนัด (PD) ที่ตรงกับความจริง**
3. **10.3** กำหนด **เกณฑ์อนุมัติทางธุรกิจ** จาก % ผิดนัด เทียบกับกำไร/ขาดทุนจริงของสินเชื่อ

```python
# 10.1 Retrain ด้วย Best Params บน training set ทั้งหมด แล้วเปรียบเทียบบน validation
tr_idx = np.random.RandomState(RANDOM_STATE).choice(len(y_train), min(200_000, len(y_train)), replace=False)
X_tr_s, y_tr_s = X_train.iloc[tr_idx], y_train[tr_idx]      # ตัวอย่าง train สำหรับวัด train ROC-AUC (Diagnosis)

def retrain_full(name, strategy):
    X, y = X_train, y_train
    cap = FIT_CAP.get(name)
    if cap and len(y) > cap:
        X, _, y, _ = train_test_split(X, y, train_size=cap, stratify=y, random_state=RANDOM_STATE)
    return fit_with_strategy(search_results[name].best_estimator_, X, y, strategy), len(y)

tuned_models, val_rows = {}, []
for name in TOP3:
    t0 = time.time()
    # ใช้ BEST_STRAT[name] อย่างถูกต้อง (ซึ่งสำหรับ HistGradientBoosting จะเป็น 'smote' ตามผลใน 9.2)
    strategy = BEST_STRAT[name]
    tuned_models[name], n_fit = retrain_full(name, strategy)
    p_val = tuned_models[name].predict_proba(X_val)[:, 1]

    # วัดค่า ROC-AUC บน Training Set ตัวอย่าง
    p_tr = tuned_models[name].predict_proba(X_tr_s)[:, 1]

    val_rows.append({
        'model': name,
        'imbalance': strategy,
        'train rows (ก่อน resample)': n_fit,
        'train ROC-AUC': roc_auc_score(y_tr_s, p_tr),
        'val ROC-AUC': roc_auc_score(y_val, p_val),
        'val PR-AUC': average_precision_score(y_val, p_val),
        'fit time (s)': round(time.time() - t0)
    })

tuned_val_df = pd.DataFrame(val_rows).set_index('model').sort_values('val ROC-AUC', ascending=False)
display(tuned_val_df)

BEST_NAME = tuned_val_df.index[0]
IMB_STRATEGY = BEST_STRAT[BEST_NAME]
final_model = tuned_models[BEST_NAME]
FINAL_NAME = f'{BEST_NAME} ({IMB_STRATEGY})'
p_val_final = final_model.predict_proba(X_val)[:, 1]
print(f'โมเดลสุดท้าย (val ROC-AUC สูงสุด): {FINAL_NAME} | val ROC-AUC = {roc_auc_score(y_val, p_val_final):.4f}')
```

### 10.2 Calibration: แปลงคะแนนของโมเดลเป็น % ผิดนัด (ใช้ validation set)
คะแนนดิบของโมเดล (โดยเฉพาะเมื่อใช้ class_weight / resampling หรือโมเดลต้นไม้) **จัดลำดับความเสี่ยงได้ แต่ไม่ใช่ % ที่ตรงกับความจริง** — เช่น ถ้าเทรนด้วย undersampling โมเดลเห็นผู้ผิดนัด 50% จึงให้คะแนนสูงเกินจริง

**Platt scaling:** fit Logistic Regression จาก logit(คะแนนดิบ) → ผลจริง บน validation set แล้วใช้แปลงคะแนนของลูกค้าทุกรายเป็น **PD (%)** — เป็นฟังก์ชันเพิ่มขึ้นทางเดียว ลำดับความเสี่ยงจึงไม่เปลี่ยน (ROC-AUC เท่าเดิม) ตรวจผลบน test ในหัวข้อ 11.2

```python
def _logit(p):
    p = np.clip(p, 1e-6, 1 - 1e-6); return np.log(p / (1 - p)).reshape(-1, 1)
calibrator = LogisticRegression().fit(_logit(p_val_final), y_val)
def predict_pd(model, X):
    """% ผิดนัด (0–1) ของลูกค้าแต่ละราย"""
    return calibrator.predict_proba(_logit(model.predict_proba(X)[:, 1]))[:, 1]

pd_val = calibrator.predict_proba(_logit(p_val_final))[:, 1]
display(pd.DataFrame({'mean predicted (%)': [p_val_final.mean() * 100, pd_val.mean() * 100],
                      'actual default (%)': [y_val.mean() * 100] * 2,
                      'Brier score': [brier_score_loss(y_val, p_val_final), brier_score_loss(y_val, pd_val)]},
                     index=['คะแนนดิบ (ก่อน calibrate)', 'PD (หลัง calibrate)']))
```

### 10.3 เกณฑ์อนุมัติทางธุรกิจ: อนุมัติเมื่อ "กำไรที่คาดหวัง" เป็นบวก
เมื่อรู้ % ผิดนัด (PD) ของลูกค้าแต่ละราย จะคำนวณกำไรที่คาดหวังของสินเชื่อได้

$$E[\text{กำไร}] = (1-\text{PD})\times g \times \text{วงเงิน} \;-\; \text{PD}\times l \times \text{วงเงิน} > 0 \;\;\Longleftrightarrow\;\; \text{PD} < \frac{g}{g+l}\;\;(\text{จุดคุ้มทุน})$$

* $g$ = ผลตอบแทนเฉลี่ยต่อวงเงินของสินเชื่อที่ **ชำระครบ** = (`total_pymnt` − `loan_amnt`) / `loan_amnt`
* $l$ = ผลขาดทุนเฉลี่ยต่อวงเงินของสินเชื่อที่ **ผิดนัด** = (`loan_amnt` − `total_pymnt`) / `loan_amnt` (รวมดอกเบี้ยและเงินที่ติดตามคืนได้แล้ว)
* คำนวณ $g, l$ **จาก train set เท่านั้น** แยกตามระยะกู้ (36 / 60 เดือน) เพราะผลตอบแทนต่างกันมาก
* `total_pymnt` เป็นข้อมูลหลังเหตุการณ์ → ใช้ **ประเมินมูลค่าทางธุรกิจเท่านั้น ไม่ใช่ feature**
* ข้อดีเทียบการสมมติต้นทุน FN : FP: เกณฑ์มาจาก **กำไร/ขาดทุนจริงของพอร์ต** อธิบายผู้บริหารได้ตรง ๆ ว่า "อนุมัติเมื่อคาดว่าได้กำไร"

ตรวจบน validation: เทียบกำไรจริงของนโยบายนี้กับ "อนุมัติทุกราย" และกับการลองทุก threshold

```python
ret_tr = (train_df['total_pymnt'] - train_df['loan_amnt']) / train_df['loan_amnt']
econ = pd.DataFrame({
    'g (ผลตอบแทนเมื่อชำระครบ)': ret_tr[train_df[TARGET] == 0].groupby(train_df['term_months']).mean(),
    'l (ขาดทุนเมื่อผิดนัด)': (-ret_tr[train_df[TARGET] == 1]).groupby(train_df['term_months']).mean(),
})
econ['จุดคุ้มทุน PD = g/(g+l)'] = econ.iloc[:, 0] / (econ.iloc[:, 0] + econ.iloc[:, 1])
display((econ * 100).round(2).rename_axis('term (months)').add_suffix(' (%)'))
BREAK_EVEN = econ['จุดคุ้มทุน PD = g/(g+l)'].to_dict()
DEFAULT_BE = float(econ['จุดคุ้มทุน PD = g/(g+l)'].mean())

def pd_threshold(term_months):
    return np.array([BREAK_EVEN.get(t, DEFAULT_BE) for t in np.asarray(term_months)])

def decide(pd_, term_months):
    """1 = REVIEW/REJECT (กำไรคาดหวัง ≤ 0), 0 = APPROVE"""
    return (np.asarray(pd_) >= pd_threshold(term_months)).astype(int)

def policy_report(df_part, pd_, label):
    profit = np.nan_to_num((df_part['total_pymnt'] - df_part['loan_amnt']).values)
    flag = decide(pd_, df_part['term_months'].values); appr = flag == 0; y_ = df_part[TARGET].values
    return {'นโยบาย': label, 'อนุมัติ (%)': appr.mean() * 100,
            'default rate ของที่อนุมัติ (%)': y_[appr].mean() * 100 if appr.any() else np.nan,
            'กำไรจริงรวม ($M)': profit[appr].sum() / 1e6,
            'ขาดทุนที่หลีกเลี่ยงได้ ($M)': -profit[~appr & (y_ == 1)].sum() / 1e6,
            'กำไรที่เสียโอกาส ($M)': profit[~appr & (y_ == 0)].sum() / 1e6}

pol_val = pd.DataFrame([
    {'นโยบาย': 'อนุมัติทุกราย (ไม่ใช้โมเดล)', 'อนุมัติ (%)': 100.0, 'default rate ของที่อนุมัติ (%)': y_val.mean() * 100,
     'กำไรจริงรวม ($M)': (val_df['total_pymnt'] - val_df['loan_amnt']).sum() / 1e6,
     'ขาดทุนที่หลีกเลี่ยงได้ ($M)': 0.0, 'กำไรที่เสียโอกาส ($M)': 0.0},
    policy_report(val_df, pd_val, 'โมเดล: อนุมัติเมื่อ PD < จุดคุ้มทุน'),
]).set_index('นโยบาย')
display(pol_val)

# ตรวจสอบ: กำไรจริงบน validation ถ้าใช้ threshold PD เดียวทุกระยะกู้
val_profit = (val_df['total_pymnt'] - val_df['loan_amnt']).values
grid_t = np.linspace(0.05, 0.8, 76)
prof_curve = [val_profit[pd_val < t].sum() / 1e6 for t in grid_t]
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(grid_t * 100, prof_curve, label='single PD threshold')
for t, v in BREAK_EVEN.items():
    ax.axvline(v * 100, ls='--', c='r' if t == 36 else 'orange', label=f'break-even PD ({int(t)} months) = {v:.1%}')
ax.axhline(val_profit.sum() / 1e6, ls=':', c='k', label='approve all')
ax.set_xlabel('approve if PD < threshold (%)'); ax.set_ylabel('realized profit on validation ($M)')
ax.set_title('Profit vs PD threshold (validation)'); ax.legend(fontsize=8); plt.tight_layout(); plt.show()
print(f'threshold เดียวที่กำไรสูงสุดบน validation ≈ {grid_t[int(np.argmax(prof_curve))]:.1%} '
      f'(เทียบจุดคุ้มทุนจาก train: {", ".join(f"{int(k)} เดือน = {v:.1%}" for k, v in BREAK_EVEN.items())})')

# ระดับความเสี่ยง (risk band) สำหรับใช้งาน — ขอบเขตกำหนดจาก PD
BANDS = [0, 0.10, 0.20, 0.30, 0.45, 1.0]
BAND_LABELS = ['1-ต่ำมาก (<10%)', '2-ต่ำ (10–20%)', '3-ปานกลาง (20–30%)', '4-สูง (30–45%)', '5-สูงมาก (≥45%)']
BAND_ACTION = {BAND_LABELS[0]: 'อนุมัติอัตโนมัติ', BAND_LABELS[1]: 'อนุมัติ',
               BAND_LABELS[2]: 'อนุมัติตามเกณฑ์คุ้มทุน / ลดวงเงิน', BAND_LABELS[3]: 'ส่งตรวจสอบเพิ่ม / ขอหลักประกัน',
               BAND_LABELS[4]: 'ปฏิเสธหรือเสนอวงเงิน/ระยะสั้นลง'}
def risk_band(pd_):
    return pd.cut(np.asarray(pd_), BANDS, labels=BAND_LABELS, right=False, include_lowest=True)
```

### 11.2 % ผิดนัดที่ทำนายแม่นแค่ไหน (Calibration) — คำตอบของคำถามโดยตรง
* **Calibration curve:** แบ่งลูกค้าใน test เป็น 10 กลุ่มตาม % ที่ทำนาย (decile) แล้วเทียบ **% ผิดนัดที่ทำนายเฉลี่ย** กับ **% ผิดนัดจริง** — ถ้าแม่น จุดจะอยู่บนเส้นทแยง
* **MAE / RMSE (จุด%)** ของ (ทำนาย − จริง) รายกลุ่ม: error หน่วยเดียวกับคำตอบ เข้าใจง่ายสำหรับผู้บริหาร
* **Brier score / Log loss** ระดับรายบุคคล เทียบกับ baseline ที่ให้ทุกคน = อัตราผิดนัดเฉลี่ยของ train

```python
def log_loss_(y, p):
    p = np.clip(p, 1e-15, 1 - 1e-15); return float(-np.mean(y * np.log(p) + (1 - y) * np.log(1 - p)))

p_base = np.full_like(pd_test, y_train.mean())
p_raw_cal_check = np.clip(p_test, 1e-6, 1 - 1e-6)
cal_metrics = pd.DataFrame({
    'Brier score': [brier_score_loss(y_test, p_base), brier_score_loss(y_test, p_raw_cal_check), brier_score_loss(y_test, pd_test)],
    'Log loss': [log_loss_(y_test, p_base), log_loss_(y_test, p_raw_cal_check), log_loss_(y_test, pd_test)],
    'mean predicted %': [p_base.mean() * 100, p_test.mean() * 100, pd_test.mean() * 100],
    'actual default %': [y_test.mean() * 100] * 3,
}, index=['baseline (ทุกคน = อัตราเฉลี่ย)', f'{FINAL_NAME} คะแนนดิบ (ก่อน calibrate)', f'{FINAL_NAME} PD (หลัง calibrate)'])
display(cal_metrics)

# Calibration table รายกลุ่ม (decile ของ % ที่ทำนาย)
cal = pd.DataFrame({'PD': pd_test, 'y': y_test})
cal['decile'] = pd.qcut(cal['PD'].rank(method='first'), 10, labels=range(1, 11))
cal_tbl = cal.groupby('decile', observed=True).agg(n=('y', 'size'), pred_pct=('PD', 'mean'), actual_pct=('y', 'mean'),
                                                   min_pred=('PD', 'min'), max_pred=('PD', 'max'))
cal_tbl[['pred_pct', 'actual_pct', 'min_pred', 'max_pred']] *= 100
cal_tbl['error (จุด%)'] = cal_tbl['pred_pct'] - cal_tbl['actual_pct']
display(cal_tbl.rename(columns={'pred_pct': '% ผิดนัดที่ทำนาย (เฉลี่ย)', 'actual_pct': '% ผิดนัดจริง',
                                'min_pred': 'PD ต่ำสุด (%)', 'max_pred': 'PD สูงสุด (%)'}))
CAL_MAE = cal_tbl['error (จุด%)'].abs().mean(); CAL_RMSE = np.sqrt((cal_tbl['error (จุด%)'] ** 2).mean())
print(f'Calibration error รายกลุ่ม: MAE = {CAL_MAE:.2f} จุด% | RMSE = {CAL_RMSE:.2f} จุด%')

fig, ax = plt.subplots(1, 2, figsize=(14, 5))
raw_cal = pd.DataFrame({'p': p_test, 'y': y_test})
raw_cal['d'] = pd.qcut(raw_cal['p'].rank(method='first'), 10, labels=False)
rc_ = raw_cal.groupby('d').mean() * 100
ax[0].plot(rc_['p'], rc_['y'], 'o--', c='gray', label='raw score (before calibration)')
ax[0].plot(cal_tbl['pred_pct'], cal_tbl['actual_pct'], 'o-', label='PD (after calibration)')
lim = [0, max(rc_['p'].max(), cal_tbl['actual_pct'].max()) * 1.05]
ax[0].plot(lim, lim, 'k:', label='perfect calibration')
ax[0].set_xlabel('predicted default % (decile mean)'); ax[0].set_ylabel('actual default % (decile)')
ax[0].set_title('Calibration curve (test)'); ax[0].legend()
ax[1].hist(pd_test * 100, bins=50, color='#4C72B0')
ax[1].set_xlabel('predicted probability of default (%)'); ax[1].set_ylabel('number of customers')
ax[1].set_title('Distribution of predicted default % (test)')
plt.tight_layout(); plt.show()

# ตัวอย่างคำตอบ: % ผิดนัดของลูกค้าแต่ละราย (test set)
show = test_df[['id', 'grade', 'loan_amnt', 'term_months', 'annual_inc', 'fico_avg', 'dti']].copy()
show['% ผิดนัด (PD)'] = (pd_test * 100).round(1)
show['ระดับความเสี่ยง'] = risk_band(pd_test)
show['คำแนะนำ'] = np.where(pred_test == 1, 'REVIEW/REJECT', 'APPROVE')
show['ผลจริง'] = np.where(y_test == 1, 'Charged Off', 'Fully Paid')
display(show.sample(10, random_state=RANDOM_STATE))
display(pd.DataFrame({'% ผิดนัดเฉลี่ย (PD)': show.groupby('grade', observed=True)['% ผิดนัด (PD)'].mean(),
                      '% ผิดนัดจริง': test_df.groupby('grade', observed=True)[TARGET].mean() * 100}))
```

## 12. Diagnosis
### 12.1 เทียบผล train vs. Validation (Overfitting / Underfitting)
| ผล | การวินิจฉัย | สิ่งที่ควรทำ |
|---|---|---|
| train **สูง** / test **ต่ำ** (gap > 0.05 AUC) | **Overfitting** | ลดความซับซ้อน (max_depth, min_samples_leaf), regularization, เพิ่มข้อมูล |
| train **ต่ำ** / test **ต่ำ** (test AUC < 0.6) | **Underfitting** | เพิ่ม feature / ใช้โมเดลที่ซับซ้อนขึ้น |
| train ≈ CV ≈ validation ≈ test | **Good fit** | – |

```python
# ตารางเปรียบเทียบ Train / Cross-Validation (mean ± SD) / Validation ของ 3 โมเดลที่จูนแล้ว และ baselines เพื่อดู Overfitting / Underfitting
comp = []

# 1. 3 โมเดลที่ได้รับการจูน (Tuned Models)
for name, m in tuned_models.items():
    s = tuned_cv_df.loc[name]
    is_final = (name == BEST_NAME)
    setting_name = 'FINAL MODEL' if is_final else 'Tuned (Full Train)'

    comp.append({
        'model': f'{name} [Tuned]',
        'setting': setting_name,
        'train ROC-AUC': tuned_val_df.loc[name, 'train ROC-AUC'],
        'CV ROC-AUC (mean ± SD)': f"{s['CV ROC-AUC mean']:.4f} ± {s['CV ROC-AUC SD']:.4f}",
        'val ROC-AUC': tuned_val_df.loc[name, 'val ROC-AUC'],
        'test ROC-AUC': roc_auc_score(y_test, m.predict_proba(X_test)[:, 1])
    })

# 2. Baseline: sub_grade อย่างเดียว (LR)
comp.append({
    'model': 'sub_grade only (LR)',
    'setting': 'baseline (เกรดของ Lending Club)',
    'train ROC-AUC': np.nan,
    'CV ROC-AUC (mean ± SD)': '–',
    'val ROC-AUC': roc_auc_score(y_val, sg_model.predict_proba(val_df[['sub_grade_num']])[:, 1]),
    'test ROC-AUC': roc_auc_score(y_test, p_sg)
})

# 3. Baseline: Dummy (prior)
comp.append({
    'model': 'Dummy (prior)',
    'setting': 'baseline (เดาสุ่มตามสัดส่วนคลาส)',
    'train ROC-AUC': 0.5000,
    'CV ROC-AUC (mean ± SD)': '0.5000 ± 0.0000',
    'val ROC-AUC': 0.5000,
    'test ROC-AUC': 0.5000
})

comp_df = pd.DataFrame(comp).set_index('model')

# คำนวณช่องว่างความแตกต่าง (Gap) ระหว่าง Train กับ Validation เพื่อวิเคราะห์ Overfitting
comp_df['train − val gap'] = comp_df['train ROC-AUC'] - comp_df['val ROC-AUC']
comp_df['train − test gap'] = comp_df['train ROC-AUC'] - comp_df['test ROC-AUC']
comp_df['diagnosis'] = np.where(comp_df['train − val gap'] > 0.05, 'Overfitting (gap > 0.05)',
                         np.where(comp_df['val ROC-AUC'] < 0.6, 'Underfitting / Weak Model',
                         np.where(comp_df.index == 'Dummy (prior)', 'Baseline', 'Good Fit')))

display(comp_df.sort_values('val ROC-AUC', ascending=False))

# ---------- Training Loss Curve: Train vs Validation ตามจำนวนต้นไม้ (Gradient Boosting) ----------
from sklearn.metrics import log_loss

gb_name = BEST_NAME if BEST_NAME in ('HistGradientBoosting', 'XGBoost') else \
          next((n for n in ('HistGradientBoosting', 'XGBoost') if n in tuned_models), None)

if gb_name is None:
    print('ไม่มีโมเดลแบบ Gradient Boosting ใน TOP3 จึงวาด loss curve ไม่ได้')
else:
    gb_pipe = tuned_models[gb_name]
    prep_gb, clf_gb = gb_pipe.named_steps['prep'], gb_pipe.named_steps['model']
    Z_tr, Z_val = prep_gb.transform(X_tr_s), prep_gb.transform(X_val)   # X_tr_s = ตัวอย่าง train 200,000 แถว (หัวข้อ 10.1)

    def _rmse(y, p): return float(np.sqrt(np.mean((y - p) ** 2)))     # RMSE ของความน่าจะเป็น = sqrt(Brier score)

    if gb_name == 'HistGradientBoosting':
        stages = zip(clf_gb.staged_predict_proba(Z_tr), clf_gb.staged_predict_proba(Z_val))
        tr_rmse, va_rmse, tr_ll, va_ll = [], [], [], []
        for p_tr_, p_va_ in stages:
            tr_rmse.append(_rmse(y_tr_s, p_tr_[:, 1])); va_rmse.append(_rmse(y_val, p_va_[:, 1]))
            tr_ll.append(log_loss(y_tr_s, p_tr_[:, 1])); va_ll.append(log_loss(y_val, p_va_[:, 1]))
    else:   # XGBoost: ทำนายด้วยต้นไม้ 1..k ต้น
        n_trees = clf_gb.get_booster().num_boosted_rounds()
        tr_rmse, va_rmse, tr_ll, va_ll = [], [], [], []
        for k in range(1, n_trees + 1):
            p_tr_ = clf_gb.predict_proba(Z_tr, iteration_range=(0, k))[:, 1]
            p_va_ = clf_gb.predict_proba(Z_val, iteration_range=(0, k))[:, 1]
            tr_rmse.append(_rmse(y_tr_s, p_tr_)); va_rmse.append(_rmse(y_val, p_va_))
            tr_ll.append(log_loss(y_tr_s, p_tr_)); va_ll.append(log_loss(y_val, p_va_))

    trees = np.arange(1, len(tr_rmse) + 1)
    fig, ax = plt.subplots(1, 2, figsize=(16, 5.5))
    for a, tr_, va_, lab in [(ax[0], tr_rmse, va_rmse, 'RMSE'), (ax[1], tr_ll, va_ll, 'Log loss')]:
        a.plot(trees, tr_, color='#3B8ED0', lw=1.6, label='Train Loss')
        a.plot(trees, va_, color='#D9534F', lw=1.6, label='Val Loss')
        best = int(np.argmin(va_))
        a.axvline(trees[best], ls=':', c='gray')
        a.set_xlabel('Number of Trees'); a.set_ylabel(lab); a.legend()
        top = max(max(tr_), max(va_))
        a.set_ylim(0, top * 1.1)                       # แกน y เริ่มที่ 0 ให้เห็นระยะห่างตามจริง
        a.set_xlim(0, len(trees) + 1)
    plt.tight_layout(); plt.show()

    best = int(np.argmin(va_ll))
    print(f'จำนวนต้นไม้ทั้งหมด = {len(trees)} | Val log loss ต่ำสุดที่ต้นที่ {trees[best]}')
    print(f'ต้นสุดท้าย: Train RMSE = {tr_rmse[-1]:.4f}, Val RMSE = {va_rmse[-1]:.4f} '
          f'(gap = {va_rmse[-1] - tr_rmse[-1]:+.4f}) | Train log loss = {tr_ll[-1]:.4f}, Val log loss = {va_ll[-1]:.4f}')
    print('อ่านผล: เส้น Val ลดลงตาม Train และไม่เด้งขึ้นตอนท้าย → ไม่ overfit; '
          'ถ้า Val เริ่มสูงขึ้นขณะ Train ยังลด → overfit (ควรใช้จำนวนต้นไม้ ณ เส้นประ)')
```

### 12.1.1 ตรวจสอบ Overfitting เฉพาะโมเดลที่เลือก (Final Model)

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import roc_auc_score

# ============================================================
# Learning Curve: HistGradientBoosting
# X-axis = Boosting Iterations
# Y-axis = ROC-AUC
# Lines = Train and Validation
# ============================================================

# Update to match the exact key stored in tuned_models ('HistGradientBoosting')
selected_model_name = 'HistGradientBoosting'

# เลือกโมเดลที่จูนแล้ว
if selected_model_name not in tuned_models:
    print(f'ไม่พบ {selected_model_name} ใน tuned_models (ไม่ได้อยู่ใน TOP3: {TOP3}) — ข้ามกราฟนี้')
else:

    model = clone(tuned_models[selected_model_name])

    # ดึง HistGradientBoostingClassifier ออกจาก Pipeline
    if isinstance(model, Pipeline):
        clf_name = next(
            (
                name for name, step in reversed(model.steps)
                if isinstance(step, HistGradientBoostingClassifier)
            ),
            None
        )

        if clf_name is None:
            raise TypeError(
                'ไม่พบ HistGradientBoostingClassifier ใน Pipeline'
            )

        clf = model.named_steps[clf_name]

        # ฝึก Pipeline ให้ fit preprocessing และโมเดลจาก Train เท่านั้น
        model.fit(X_train, y_train)

        # แปลงข้อมูลด้วย preprocessing ที่ fit จาก Train
        prep_steps = model.steps[:model.steps.index((clf_name, clf))]
        if prep_steps:
            prep_lc = Pipeline(prep_steps)          # ไม่ใช้ชื่อ preprocess เพื่อไม่ทับตัวแปรหลักจากหัวข้อ 6
            X_train_transformed = prep_lc.transform(X_tr_s)
            X_val_transformed = prep_lc.transform(X_val)
        else:
            X_train_transformed = X_tr_s
            X_val_transformed = X_val

    else:
        if not isinstance(model, HistGradientBoostingClassifier):
            raise TypeError(
                'โมเดลที่เลือกไม่ใช่ HistGradientBoostingClassifier'
            )

        clf = model
        clf.fit(X_train, y_train)
        X_train_transformed = X_tr_s
        X_val_transformed = X_val

    # คำนวณ ROC-AUC ในแต่ละรอบ
    train_auc = []
    val_auc = []

    for train_prob, val_prob in zip(
        clf.staged_predict_proba(X_train_transformed),
        clf.staged_predict_proba(X_val_transformed)
    ):
        train_auc.append(
            roc_auc_score(y_tr_s, train_prob[:, 1])
        )
        val_auc.append(
            roc_auc_score(y_val, val_prob[:, 1])
        )

    iterations = np.arange(1, len(train_auc) + 1)

    # วาดกราฟ
    plt.figure(figsize=(10, 6))

    plt.plot(
        iterations,
        train_auc,
        label='Train ROC-AUC',
        linewidth=2
    )

    plt.plot(
        iterations,
        val_auc,
        label='Validation ROC-AUC',
        linewidth=2
    )

    plt.xlabel('Boosting Iterations')
    plt.ylabel('ROC-AUC Score')
    plt.title('HistGradientBoosting: Train vs Validation Learning Curve')
    plt.ylim(0.45, 1.00)
    plt.grid(axis='both', linestyle='--', alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()

    # แสดงรอบที่ Validation ROC-AUC สูงที่สุด
    best_idx = int(np.argmax(val_auc))

    print(f'จำนวนรอบที่ Validation ROC-AUC สูงที่สุด: {iterations[best_idx]}')
    print(f'Train ROC-AUC:      {train_auc[best_idx]:.4f}')
    print(f'Validation ROC-AUC: {val_auc[best_idx]:.4f}')
```

* **Monitoring – Model Decay:** ติดตาม ROC-AUC ตามช่วงเวลา (ปีที่ออกสินเชื่อ) ถ้าลดลงต่ำกว่าเกณฑ์ (เช่น ต่ำกว่า validation เกิน 0.03) ให้เตรียม retrain

```python
# Save & reload model
artifact = {'model': final_model, 'calibrator': calibrator, 'features': FEATURES,
            'break_even_pd': BREAK_EVEN, 'default_break_even': DEFAULT_BE,
            'risk_bands': BANDS, 'band_labels': BAND_LABELS, 'band_action': BAND_ACTION,
            'model_name': FINAL_NAME, 'sklearn_version': sklearn.__version__}
joblib.dump(artifact, 'lending_club_default_model.joblib', compress=3)
loaded = joblib.load('lending_club_default_model.joblib')
sample_apps = X_test.head(5)
raw_score = loaded['model'].predict_proba(sample_apps[loaded['features']])[:, 1]
pd_score = loaded['calibrator'].predict_proba(_logit(raw_score))[:, 1]
thr5 = np.array([loaded['break_even_pd'].get(t, loaded['default_break_even']) for t in test_df['term_months'].head(5)])
display(pd.DataFrame({'% ผิดนัด (PD)': (pd_score * 100).round(1), 'decision': np.where(pd_score >= thr5, 'REVIEW/REJECT', 'APPROVE'),
                      'actual': y_test[:5]}, index=sample_apps.index))

# ตัวอย่างฟังก์ชันสำหรับ API / application
def score_application(app: dict, term_months: int, art=loaded):
    # รับข้อมูลใบสมัคร 1 ราย -> % ผิดนัด, ระดับความเสี่ยง, คำแนะนำ
    X_new = pd.DataFrame([app])[art['features']]
    pd_ = float(art['calibrator'].predict_proba(_logit(art['model'].predict_proba(X_new)[:, 1]))[0, 1])
    thr = art['break_even_pd'].get(term_months, art['default_break_even'])
    band = str(pd.cut([pd_], art['risk_bands'], labels=art['band_labels'], right=False, include_lowest=True)[0])
    return {'% ผิดนัด': round(pd_ * 100, 1), 'ระดับความเสี่ยง': band, 'จุดคุ้มทุน (%)': round(thr * 100, 1),
            'คำแนะนำ': 'REVIEW/REJECT' if pd_ >= thr else 'APPROVE', 'แนวทาง': art['band_action'][band]}
print('score_application(ใบสมัครตัวอย่าง) ->', score_application(X_test.iloc[0].to_dict(), int(test_df['term_months'].iloc[0])))
print(f'ขนาดไฟล์โมเดล: {os.path.getsize("lending_club_default_model.joblib")/1024**2:.1f} MB')

# Data drift: PSI
def psi(expected, actual, bins=10):
    expected, actual = pd.Series(expected).dropna(), pd.Series(actual).dropna()
    if expected.nunique() <= bins:                       # ตัวแปรที่มีไม่กี่ค่า (เช่น term) ใช้สัดส่วนรายค่า
        cats = sorted(set(expected) | set(actual))
        e = expected.value_counts(normalize=True).reindex(cats, fill_value=0).values
        a = actual.value_counts(normalize=True).reindex(cats, fill_value=0).values
        e, a = np.clip(e, 1e-6, None), np.clip(a, 1e-6, None)
        return float(np.sum((a - e) * np.log(a / e)))
    edges = np.unique(np.quantile(expected, np.linspace(0, 1, bins + 1)))
    if len(edges) < 3: return np.nan
    e = np.histogram(np.clip(expected, edges[0], edges[-1]), edges)[0] / len(expected)
    a = np.histogram(np.clip(actual, edges[0], edges[-1]), edges)[0] / len(actual)
    e, a = np.clip(e, 1e-6, None), np.clip(a, 1e-6, None)
    return float(np.sum((a - e) * np.log(a / e)))

base = train_df[train_df['issue_year'] <= 2015]
new = pd.concat([val_df, test_df])
new = new[new['issue_year'] == 2016]      # ข้อมูลใหม่จำลอง (ไม่ใช้ train)
drift_cols = imp['feature'].head(8).tolist()
drift_cols = [c for c in drift_cols if c in NUM_SEL]
psi_rows = [{'feature': c, 'PSI': psi(base[c], new[c])} for c in drift_cols]
psi_rows.append({'feature': 'model score', 'PSI': psi(final_model.predict_proba(base[FEATURES].head(200_000))[:, 1],
                                                        final_model.predict_proba(new[FEATURES].head(200_000))[:, 1])})
psi_df = pd.DataFrame(psi_rows).set_index('feature')
psi_df['status'] = pd.cut(psi_df['PSI'], [-np.inf, 0.1, 0.25, np.inf], labels=['stable', 'moderate shift', 'significant shift'])
display(psi_df)

# Model decay: ROC-AUC ของโมเดลสุดท้ายแยกตามปี (test set)
decay = pd.DataFrame({'year': test_df['issue_year'].values, 'y': y_test, 'p': p_test}) \
          .groupby('year').apply(lambda d: pd.Series({'n': len(d), 'ROC-AUC': roc_auc_score(d['y'], d['p']) if d['y'].nunique() > 1 else np.nan}))
decay['ต่ำกว่า val เกิน 0.03?'] = decay['ROC-AUC'] < roc_auc_score(y_val, p_val_final) - 0.03
display(decay)
fig, ax = plt.subplots(figsize=(8, 4))
decay['ROC-AUC'].plot(marker='o', ax=ax); ax.axhline(roc_auc_score(y_val, p_val_final), ls='--', c='k', label='validation AUC')
ax.set_title('Model decay check: test ROC-AUC by issue year'); ax.legend(); plt.tight_layout(); plt.show()
```

## สรุปผล (Conclusion)
**คำถาม: ลูกค้าแต่ละรายมีโอกาสผิดนัดชำระหนี้กี่เปอร์เซ็นต์ โดยพิจารณาจากข้อมูล ณ วันที่สมัครขอสินเชื่อ?**

* **คำตอบ:** โมเดล **HistGradientBoosting** (retrain ด้วย Best Params บน train ทั้งหมด) + Platt calibration ให้ **% ผิดนัด (PD) ของลูกค้าแต่ละราย** ได้ (ตัวอย่างในหัวข้อ 11.2 และฟังก์ชัน `score_application()` ในหัวข้อ 13) — PD ของลูกค้าใน test อยู่ในช่วงประมาณ **0.9% ถึง 73.8%** (เฉลี่ย **19.7%** เทียบกับอัตราผิดนัดจริง **19.7%**)
* **% ที่ทำนายเชื่อถือได้ระดับกลุ่ม:** calibration curve อยู่ใกล้เส้นทแยง — error ระหว่าง % ที่ทำนายกับ % ผิดนัดจริงรายกลุ่ม (decile) MAE = **0.21 จุด%** และ Brier score = **0.1419** (baseline **0.1583**) เช่น กลุ่มที่โมเดลบอก ~**47.3%** ผิดนัดจริง **47.1%** และกลุ่มที่โมเดลบอก ~**3.9%** ผิดนัดจริง **3.8%**
* **จัดลำดับความเสี่ยงได้ในระดับปานกลาง:** ROC-AUC = **0.7206** บน test ดีกว่าการเดา (0.50) และดีกว่าใช้เกรดของ Lending Club อย่างเดียว (**0.6881**) — กลุ่ม 10% ที่เสี่ยงต่ำสุดผิดนัดจริง **3.8%** ส่วนกลุ่ม 10% ที่เสี่ยงสูงสุดผิดนัดจริง **47.1%**
* **ปัจจัยที่ทำให้ % ผิดนัดสูงขึ้นมากที่สุด** (permutation importance + EDA): **sub_grade** (เกรดความเสี่ยงของ Lending Club ซึ่งสัมพันธ์กับดอกเบี้ยสูงมาก), **ระยะกู้ 60 เดือน (term)**, **FICO ต่ำ** รองลงมาคือ dti สูงและวงเงินกู้ต่อรายได้สูง
* **ข้อควรเข้าใจ:** PD เป็น **ค่าความน่าจะเป็น** ไม่ใช่คำทำนายแน่นอน — ลูกค้า PD 30% หมายถึง ในลูกค้าลักษณะเดียวกัน 100 ราย คาดว่าผิดนัดราว 30 ราย; การผิดนัดยังขึ้นกับเหตุการณ์หลังอนุมัติ (ตกงาน เจ็บป่วย เศรษฐกิจ) ที่ข้อมูล ณ วันสมัครไม่มี

**ผลลัพธ์นำไปใช้ประโยชน์ได้อย่างไร และโดยใคร**

* **Credit Risk Analysts / Underwriters:** ใช้ % ผิดนัดและระดับความเสี่ยง (risk band 1–5) ของผู้สมัครแต่ละรายช่วยตัดสินใจ — อนุมัติเมื่อ PD ต่ำกว่าจุดคุ้มทุน (**27.2%** สำหรับ 36 เดือน, **37.3%** สำหรับ 60 เดือน) ส่วนที่เกินส่งตรวจสอบเพิ่ม ลดวงเงิน หรือปฏิเสธ (ฟังก์ชัน `score_application()` ในหัวข้อ 13)
* **ผู้บริหาร / ฝ่ายการเงิน:** บน test set นโยบายนี้เปลี่ยนกำไรจริงจาก **126.82 ล้านดอลลาร์** (อนุมัติทุกราย) เป็น **139.48 ล้านดอลลาร์** (เพิ่มขึ้น **12.66 ล้านดอลลาร์**) และลด default rate ของสินเชื่อที่อนุมัติจาก 19.7% เหลือ 15.1% — หลีกเลี่ยงขาดทุนจากหนี้เสียได้ **92.25 ล้านดอลลาร์** แลกกับกำไรที่เสียโอกาส **79.59 ล้านดอลลาร์** **92.25 ล้านดอลลาร์** แลกกับกำไรที่เสียโอกาส **79.59 ล้านดอลลาร์**
* **Portfolio Managers:** รวม PD ของลูกค้าทั้งพอร์ต = จำนวนสินเชื่อที่คาดว่าจะผิดนัด ใช้วางแผนสำรองหนี้สูญ
* **ทีมนโยบายสินเชื่อ:** ใช้ปัจจัยที่สำคัญเป็นเกณฑ์ประกอบการพิจารณา และกำหนดเงื่อนไขตามระดับ PD

**ข้อจำกัดของข้อมูลและโมเดล**

* **Right-censoring / selection bias:** ตัดสินเชื่อที่ยังไม่สิ้นสุดและสินเชื่อหลังปี 2016 ออก แต่สินเชื่อ 60 เดือนที่ออกปี 2015–2016 ส่วนมากยังไม่จบ ส่วนที่ปิดแล้วอาจเอนไปทางผิดนัดเร็วหรือปิดก่อนกำหนด (อาจเป็นเหตุให้ default rate ปี 2016 สูงถึง 23.4%)
* **มีเฉพาะสินเชื่อที่ได้รับอนุมัติ (reject inference):** โมเดลไม่เคยเห็นผู้ที่ถูกปฏิเสธ — % ผิดนัดของผู้สมัครที่ต่างจากลูกค้าเดิมมากอาจไม่แม่น
* **Calibration ใช้ได้กับช่วงเวลาที่ใกล้เคียงข้อมูล:** ถ้าอัตราผิดนัดของตลาดเปลี่ยน (concept drift) ต้อง calibrate ใหม่ — ข้อมูลถึงปี 2018 และ Lending Club เลิกทำ P2P lending ตั้งแต่ปี 2020
* **Random split:** ข้อมูลทุกปีปนกันใน train/test ผลอาจดีกว่าการใช้งานจริงที่ทำนายอนาคต — ต้องติดตาม data drift (หัวข้อ 13)
* **จุดคุ้มทุนคำนวณจากผลตอบแทนจริงในอดีต (`total_pymnt`)** ยังไม่หักต้นทุนเงินทุน ค่าดำเนินการ และไม่ได้ปรับเรื่องการปิดหนี้ก่อนกำหนด — ธนาคารจริงควรใส่ต้นทุนเหล่านี้ ซึ่งจะทำให้จุดคุ้มทุนต่ำลง (% ผิดนัดไม่ขึ้นกับเกณฑ์นี้)
* **Fairness:** `addr_state` และตัวแปรอื่นอาจเป็นตัวแทน (proxy) ของลักษณะที่กฎหมายคุ้มครอง ต้องตรวจสอบก่อนใช้งานจริง
* **การเข้ารหัสตัวแปรหมวดหมู่:** ใช้ Label (Ordinal) Encoding ซึ่งให้ลำดับที่ไม่มีจริงกับตัวแปรอย่าง `purpose` และ `addr_state` — ไม่กระทบโมเดลต้นไม้มาก แต่อาจกระทบ Logistic Regression, LDA และ KNN
* **tuning** ใช้ตัวอย่าง 150,000 แถวจาก train และจำนวนรอบค้นหาจำกัดเพื่อให้รันบน Colab ได้ อาจยังไม่ได้ hyperparameter ที่ดีที่สุด
* **SMOTE** ทำงานบนข้อมูลที่ผ่านการ encode แล้ว จึงสร้างค่าสังเคราะห์ระหว่างรหัสของหมวดหมู่ (ไม่ใช่ค่าจริง) และไม่ได้ถูกเลือกใช้ เพราะ CV ROC-AUC ต่ำกว่าไม่ปรับ (none) ในทุกโมเดล
