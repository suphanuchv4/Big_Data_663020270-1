```python
# ----------------- Your code here -----------------
# ใบ้: อ่าน JSON lines ทีละบรรทัดด้วย generator ไม่ต้องโหลดทั้งไฟล์
# def iter_jsonl(path):
#     with open(path, encoding='utf-8') as f:
#         for line in f:
#             yield json.loads(line)

from langdetect import detect, DetectorFactory, LangDetectException

def iter_jsonl(path):
    """Stream a jsonl file line by line instead of loading it all into memory at once."""
    with open(path, encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line:
                yield json.loads(line)

FILES = [
    'posts_20241127_075347.jsonl',
    'posts_20241127_153954.jsonl',
]

records = [rec for fn in FILES for rec in iter_jsonl(fn)]
df = pd.DataFrame(records)
print(f"Loaded {len(df)} posts total")
```

```python
# คนโพสต์เมื่อไร: จำนวนโพสต์ ช่วงเวลาที่ข้อมูลครอบคลุม และชั่วโมงที่โพสต์มากที่สุด
df['created_at'] = pd.to_datetime(df['created_at'], utc=True, format='mixed', errors='coerce')
df = df.dropna(subset=['created_at']).reset_index(drop=True)
df['hour_utc'] = df['created_at'].dt.hour

print(f"Posts with valid timestamp: {len(df)}")
print(f"Date range covered: {df['created_at'].min()} to {df['created_at'].max()}")

hour_counts = df['hour_utc'].value_counts().sort_index()
print(f"Peak posting hour (UTC): {hour_counts.idxmax()}:00 ({hour_counts.max()} posts)")
```

```python
# ใช้ภาษาอะไร: รายงานภาษาหลักและสัดส่วน เพื่อบอกว่าถ้าทำคอนเทนต์ควรรองรับภาษาใด
def safe_detect(text):
    text = str(text).strip()
    if len(text) < 3:
        return None
    try:
        return detect(text)
    except LangDetectException:
        return None

df['text'] = df['text'].fillna('')
mask = df['text'].str.strip().str.len() >= 3
df.loc[mask, 'lang'] = df.loc[mask, 'text'].apply(safe_detect)

lang_counts = df['lang'].value_counts()
total_detected = lang_counts.sum()
print(f"Language detected for {total_detected}/{len(df)} posts ({total_detected/len(df)*100:.1f}%)")
print("\nTop 5 languages:")
for lang, cnt in lang_counts.head(5).items():
    print(f"  {lang}: {cnt} ({cnt/total_detected*100:.1f}%)")
```

** en = English, ja = Japanese, de = German, es = Spain, ko = Korean

```python
# คนพูดถึงอะไร: หา hashtag หรือคำ/วลีเด่นที่น่าสนใจ และยกตัวอย่างข้อความจริงประกอบ
all_text = ' '.join(df['text'].tolist())
hashtags = re.findall(
    r'#([A-Za-z0-9_\u00C0-\u024F\u3040-\u30FF\u4E00-\u9FFF]+)',
    all_text)

# ตัด hashtag ที่สั้นเกินไป เช่น #a
hashtags = [h.lower() for h in hashtags if len(h) >= 2]
hashtag_counts = Counter(hashtags)
print("Top 10 hashtags:")
for tag, cnt in hashtag_counts.most_common(10):
    pattern = rf'(?i)(?<!\w)#{re.escape(tag)}(?!\w)'
    matched = df[df['text'].str.contains(
        pattern,
        regex=True,
        na=False
    )]

    example = matched.iloc[0]['text'] if len(matched) > 0 else ""
    print(f"#{tag} (n={cnt}): {example[:150]}")
```

```python
# ข้อมูลนี้บอกอะไรได้และบอกอะไรไม่ได้: สร้าง visualization 1 รูป แล้วเขียนข้อจำกัด 3–5 บรรทัด
fig, ax = plt.subplots(figsize=(10, 4))
ax.bar(hour_counts.index, hour_counts.values, color='deepskyblue')
ax.set_xlabel('Hour (UTC)')
ax.set_ylabel('Number of posts')
ax.set_title(f'Posts by hour of day (n={len(df)})')
ax.set_xticks(range(0, 24))
plt.tight_layout()
plt.savefig('hourly_posts.png', dpi=150)
plt.show()
```
